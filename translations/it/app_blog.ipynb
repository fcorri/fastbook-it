# Creazione di un blog

Sfortunatamente, quando si tratta di bloggare, sembra che tu debba prendere una decisione difficile: utilizzare una piattaforma che semplifichi le cose ma sottoponga te e i tuoi lettori a pubblicità, paywall e tariffe, oppure trascorri ore a configurare il tuo servizio di hosting e settimane ad apprendere tutti i tipi di dettagli intricati. Forse il vantaggio più grande dell'approccio "fai da te" è che possiedi davvero i tuoi post, piuttosto che dipendere dal capriccio di un fornitore di servizi e dalle sue decisioni su come monetizzare i tuoi contenuti in futuro.

Si scopre, tuttavia, che puoi avere il meglio di entrambi i mondi!

## Bloggare con GitHub Pages

Un'ottima soluzione è ospitare il tuo blog su una piattaforma chiamata [GitHub Pages](https://pages.github.com/), che è gratuita, non ha pubblicità o pareti a pagamento e rende i tuoi dati disponibili in modo standard in modo tale che tu possa in qualsiasi momento spostare il tuo blog su un altro host. Ma tutti gli approcci che ho visto sull’utilizzo di GitHub Pages hanno richiesto la conoscenza della riga di comando e di strumenti arcani con cui probabilmente solo gli sviluppatori di software hanno familiarità. Ad esempio, la [documentazione propria](https://help.github.com/en/github/working-with-github-pages/creating-a-github-pages-site-with-jekyll) di GitHub sulla creazione di un blog include un lungo elenco di istruzioni che implicano l'installazione del linguaggio di programmazione Ruby, l'utilizzo dello strumento da riga di comando `git`, la copia dei numeri di versione e altro ancora: 17 passaggi in totale!

Per ridurre i problemi, abbiamo creato un approccio semplice che ti consente di utilizzare un'*interfaccia interamente basata su browser* per tutte le tue esigenze di blogging. Sarai attivo e funzionante con il tuo nuovo blog entro circa cinque minuti. Non costa nulla e, se lo desideri, puoi facilmente aggiungere il tuo dominio personalizzato. In questa sezione spiegheremo come farlo, utilizzando un modello che abbiamo creato chiamato *fast\_template*. (NB: assicurati di controllare il [sito web del libro](https://book.fast.ai) per gli ultimi consigli sul blog, poiché escono sempre nuovi strumenti).

### Creazione del repository

Avrai bisogno di un account su GitHub, quindi vai lì adesso e crea un account se non ne hai già uno. Assicurati di aver effettuato l'accesso. Normalmente, GitHub viene utilizzato dagli sviluppatori di software per scrivere codice e utilizzano un sofisticato strumento da riga di comando per lavorarci, ma ti mostreremo un approccio che non utilizza affatto la riga di comando!

Per iniziare, punta il browser su [https://github.com/fastai/fast_template/generate](https://github.com/fastai/fast_template/generate) (devi essere loggato su GitHub affinché il collegamento funzioni). Ciò ti consentirà di creare un luogo in cui archiviare il tuo blog, chiamato *repository*. Avrai una schermata come quella in <<github_repo>>. Tieni presente che devi inserire il nome del tuo repository utilizzando il formato *esatto* mostrato qui, ovvero il tuo nome utente GitHub seguito da `.github.io`.

<img width="440" src="../../images/fast_template/image1.png" id="github_repo" caption="Creazione del tuo repository" alt="Screenshot della pagina GitHub per la creazione di un nuovo repository">

Dopo averlo inserito e la descrizione che preferisci, fai clic su "Crea repository da modello". Puoi scegliere di rendere il repository "privato", ma dal momento che stai creando un blog che vuoi che altre persone leggano, si spera che avere i file sottostanti disponibili pubblicamente non sia un problema per te.

Ora impostiamo la tua home page!

### Impostazione della pagina iniziale

Quando i lettori arrivano al tuo blog, la prima cosa che vedranno è il contenuto di un file chiamato *index.md*. Questo è un file [markdown](https://guides.github.com/features/mastering-markdown/).  Markdown è un modo potente ma semplice per creare testo formattato, come punti elenco, corsivo, collegamenti ipertestuali e così via. È molto utilizzato, inclusa tutta la formattazione nei notebook Jupyter, quasi ogni parte del sito GitHub e molti altri luoghi su Internet. Per creare un testo di markdown, puoi semplicemente digitare un inglese semplice, quindi aggiungere alcuni caratteri speciali per aggiungere un comportamento speciale. Ad esempio, se digiti un carattere "*" prima e dopo una parola o una frase, verrà messo in *corsivo*. Proviamolo adesso.

Per aprire il file, fare clic sul relativo nome file in GitHub. Per modificarlo, fare clic sull'icona della matita all'estrema destra dello schermo, come mostrato in <<fastpage_edit>>.

<img width="800" src="../../images/fast_template/image3.png" alt="Schermata che mostra dove fare clic per modificare il file" id="fastpage_edit" caption="Fai clic su Modifica questo file">

Puoi aggiungere, modificare o sostituire i testi che vedi. Fai clic su "Anteprima modifiche" (<<fastpage_preview>>) per vedere come apparirà il testo del markdown nel tuo blog. Le linee che hai aggiunto o modificato appariranno con una barra verde sul lato sinistro.

<img width="350" src="../../images/fast_template/image4.png" alt="Schermata che mostra dove fare clic per visualizzare l'anteprima delle modifiche" id="fastpage_preview" caption="Anteprima delle modifiche per individuare eventuali errori">

Per salvare le modifiche, scorri fino alla fine della pagina e fai clic su "Conferma modifiche", come mostrato in <<fastpage_commit>>. Su GitHub, "impegnare" qualcosa significa salvarlo sul server GitHub.

<img width="600" src="../../images/fast_template/image5.png" alt="Screenshot che mostra dove fare clic per confermare le modifiche" id="fastpage_commit" caption="Conferma le modifiche per salvarle">

Successivamente, dovresti configurare le impostazioni del tuo blog. Per fare ciò, fai clic sul file chiamato *\_config.yml*, quindi fai clic sul pulsante di modifica come hai fatto per il file indice. Modifica il titolo, la descrizione e i valori del nome utente GitHub (vedi <<github_config>>. Devi lasciare i nomi prima dei due punti al loro posto e digitare i nuovi valori dopo i due punti (e uno spazio) su ogni riga. Puoi anche aggiungere il tuo indirizzo email e il nome utente Twitter se lo desideri, ma tieni presente che questi appariranno sul tuo blog pubblico se li inserisci qui.

<img width="800" src="../../images/fast_template/image6.png" id="github_config" caption="Compila il file di configurazione" alt="Screenshot che mostra il file di configurazione e come compilarlo">

Una volta terminato, conferma le modifiche proprio come hai fatto con il file indice, quindi attendi circa un minuto mentre GitHub elabora il tuo nuovo blog. Punta il tuo browser web su *&lt;username> .github.io* (sostituendo *&lt;username>* con il tuo nome utente GitHub). Dovresti vedere il tuo blog, che sarà simile a <<github_blog>>.

<img width="540" src="../../images/fast_template/image7.png" id="github_blog" caption="Il tuo blog è online!" alt="Screenshot che mostra il sito web nomeutente.github.io">

### Creazione di post

Ora sei pronto per creare il tuo primo post. Tutti i tuoi post andranno nella cartella *\_posts*. Fare clic su di esso ora, quindi fare clic sul pulsante "Crea file". Devi fare attenzione a denominare il file utilizzando il formato *&lt;anno>-&lt;mese>-&lt;giorno>-&lt;nome>.md*, come mostrato in <<fastpage_name>>, dove *&lt;anno>* è un numero di quattro cifre e *&lt;mese>* e *&lt;giorno>* sono numeri a due cifre. *&lt;nome>* può essere qualsiasi cosa tu voglia che ti aiuti a ricordare di cosa trattava questo post. L'estensione *.md* è per i documenti con ribasso.

<img width="440" src="../../images/fast_template/image8.png" alt="Schermata che mostra la sintassi corretta per creare un nuovo post sul blog" id="fastpage_name" caption="Assegnare un nome ai tuoi post">

Puoi quindi digitare il contenuto del tuo primo post. L'unica regola è che la prima riga del tuo post deve essere un titolo con ribasso. Questo viene creato inserendo `# ` all'inizio di una riga, come mostrato in <<fastpage_title>> (che crea un'intestazione di livello 1, che dovresti usare solo una volta all'inizio del documento; puoi creare intestazioni di livello 2 usando `## `, livello 3 con `###` e così via).

<img width="300" src="../../images/fast_template/image9.png" alt="Schermata che mostra l'inizio di un post del blog" id="fastpage_title" caption="Sintassi Markdown per un titolo">

Come prima, puoi fare clic sul pulsante "Anteprima" per vedere come apparirà la formattazione del markdown (<<fastpage_preview1>>).

<img width="400" src="../../images/fast_template/image10.png" alt="Schermata che mostra lo stesso post del blog interpretato in HTML" id="fastpage_preview1" caption="Come apparirà la sintassi di markdown precedente sul tuo blog">

E dovrai fare clic sul pulsante "Conferma nuovo file" per salvarlo su GitHub, come mostrato in <<fastpage_commit1>>.

<img width="700" src="../../images/fast_template/image11.png" alt="Screenshot che mostra dove fare clic per salvare il nuovo file" id="fastpage_commit1" caption="Conferma le modifiche per salvarle">

Dai di nuovo un'occhiata alla home page del tuo blog e vedrai che questo post è ora apparso--<<fastpage_live>> mostra il risultato con la posa di esempio che abbiamo appena aggiunto. (Ricorda che dovrai attendere circa un minuto affinché GitHub elabori la richiesta prima che il file venga visualizzato.)

<img width="500" src="../../images/fast_template/image12.png" alt="Schermata che mostra il primo post sul sito web del blog" id="fastpage_live" caption="Il tuo primo post è pubblicato!">

Potresti aver notato che abbiamo fornito un post di blog di esempio, che puoi procedere ed eliminare ora. Vai alla tua cartella *\_posts*, come prima, e clicca su *2020-01-14-welcome.md*. Quindi fare clic sull'icona del cestino all'estrema destra, come mostrato in <<fastpage_delete>>.

<img width="400" src="../../images/fast_template/image13.png" alt="Screenshot che mostra come eliminare il post fittizio" id="fastpage_delete" caption="Elimina il post del blog di esempio">

In GitHub, in realtà non cambia nulla finché non ti impegni, incluso quando elimini un file! Quindi, dopo aver fatto clic sull'icona del cestino, scorri fino alla fine della pagina e conferma le modifiche.

Puoi includere immagini nei tuoi post aggiungendo una riga di markdown come
quanto segue:

    ![Descrizione immagine](images/nomefile.jpg)

Perché funzioni, dovrai inserire l'immagine nella cartella *immagini*. Per fare ciò, fare clic sulla cartella *immagini*, quindi fare clic sul pulsante "Carica file" (<<fastpage_upload>>).

<img width="400" src="../../images/fast_template/image14.png" alt="Schermata che mostra come caricare nuovi file" id="fastpage_upload" caption="Carica un file dal tuo computer">

Adesso vediamo come fare tutto questo direttamente dal tuo computer.

### Sincronizzazione di GitHub e del computer

Ci sono molti motivi per cui potresti voler copiare il contenuto del tuo blog da GitHub al tuo computer: potresti voler essere in grado di leggere o modificare i tuoi post offline, o forse vorresti un backup nel caso in cui succeda qualcosa al tuo repository GitHub.

GitHub fa molto di più che consentirti semplicemente di copiare il tuo repository sul tuo computer; ti consente di *sincronizzarlo* con il tuo computer. Ciò significa che puoi apportare modifiche su GitHub e queste verranno copiate sul tuo computer e puoi apportare modifiche sul tuo computer e queste verranno copiate su GitHub. Puoi anche consentire ad altre persone di accedere e modificare il tuo blog, e le loro modifiche e le tue modifiche verranno automaticamente combinate insieme alla successiva sincronizzazione.

Per farlo funzionare, devi installare un'applicazione chiamata [GitHub Desktop](https://desktop.github.com/) sul tuo computer. Funziona su Mac, Windows e Linux. Segui le istruzioni per installarlo e quando lo esegui ti verrà chiesto di accedere a GitHub e selezionare il repository da sincronizzare. Fai clic su "Clona un repository da Internet", come mostrato in <<fastpage_clone>>.

<img src="../../images/gitblog/image1.png" width="400" alt="Uno screenshot che mostra come clonare il tuo repository" id="fastpage_clone" caption="Clona il tuo repository su GitHub Desktop">

Una volta che GitHub avrà finito di sincronizzare il tuo repository, potrai fare clic su "Visualizza i file del tuo repository in Explorer" (o Finder), come mostrato in <<fastpage_explorer>> e vedrai la copia locale del tuo blog! Prova a modificare uno dei file sul tuo computer. Quindi torna a GitHub Desktop e vedrai il pulsante "Sincronizza" in attesa che tu lo prema. Quando fai clic su di esso, le modifiche verranno copiate su GitHub, dove le vedrai riflesse sul sito web.

<img src="../../images/gitblog/image2.png" width="600" alt="Uno screenshot che mostra il repository clonato" id="fastpage_explorer" caption="Visualizzazione dei file localmente">

Se non hai mai utilizzato `git` prima, GitHub Desktop è un ottimo modo per iniziare. Come scoprirai, è uno strumento fondamentale utilizzato dalla maggior parte dei data scientist. Un altro strumento che speriamo tu apprezzerai è Jupyter Notebooks: con quello c'è anche un modo per scrivere direttamente sul tuo blog!

## Jupyter per i blog

Puoi anche scrivere post di blog utilizzando i notebook Jupyter. Le celle di markdown, le celle di codice e tutti gli output verranno visualizzati nel post del blog esportato. Il modo migliore per farlo potrebbe essere cambiato nel momento in cui stai leggendo questo libro, quindi assicurati di controllare il [sito web del libro](https://book.fast.ai) per le informazioni più recenti. Mentre scriviamo, il modo più semplice per creare un blog dai notebook è utilizzare [`fastpages`](http://fastpages.fast.ai/), che è una versione più avanzata di `fast_template`. 

Per bloggare con un taccuino, basta inserirlo nella cartella *\_notebooks* nel repository del tuo blog e verrà visualizzato nell'elenco dei post del blog. Quando scrivi sul tuo quaderno, scrivi tutto ciò che vuoi che il tuo pubblico veda. Poiché la maggior parte delle piattaforme di scrittura rende difficile includere codice e output, molti di noi hanno l'abitudine di includere meno esempi reali di quanto dovremmo. Questo è un ottimo modo per prendere l'abitudine di includere molti esempi mentre scrivi.

Spesso vorrai nascondere i boilerplate come le istruzioni di importazione. Puoi aggiungere "#hide" nella parte superiore di qualsiasi cella per fare in modo che non venga visualizzata nell'output. Jupyter visualizza il risultato dell'ultima riga di una cella, quindi non è necessario includere `print()`. (Includere codice extra che non è necessario significa che c'è un sovraccarico cognitivo maggiore per il lettore; quindi non includere codice di cui non hai realmente bisogno!)
