### Bug e ricorsi: algoritmo difettoso utilizzato per prestazioni sanitarie

The Verge ha studiato il software utilizzato in oltre la metà degli stati degli Stati Uniti per determinare quanta assistenza sanitaria ricevono le persone e ha documentato i risultati nell'articolo ["Cosa succede quando un algoritmo taglia la tua assistenza sanitaria"](https://www.theverge.com/2018/3/21/17144260/healthcare-medicaid-algorithm-arkansas-cerebral-palsy). Dopo l’implementazione dell’algoritmo in Arkansas, centinaia di persone (molte con gravi disabilità) hanno subito tagli drastici alla loro assistenza sanitaria. Ad esempio, Tammy Dobbs, una donna affetta da paralisi cerebrale che ha bisogno di un aiuto per alzarsi dal letto, andare in bagno, prendere il cibo e altro ancora, ha visto le sue ore di aiuto improvvisamente ridotte di 20 ore a settimana. Non è riuscita a ottenere alcuna spiegazione sul motivo per cui la sua assistenza sanitaria è stata tagliata. Alla fine, un caso giudiziario ha rivelato che c’erano errori nell’implementazione del software dell’algoritmo, che avevano un impatto negativo sulle persone con diabete o paralisi cerebrale. Tuttavia, Dobbs e molte altre persone che dipendono da questi benefici sanitari vivono nel timore che i loro benefici possano essere nuovamente tagliati improvvisamente e inspiegabilmente.

### Circuiti di feedback: il sistema di consigli di YouTube

Possono verificarsi cicli di feedback quando il tuo modello controlla il successivo ciclo di dati che ottieni. I dati restituiti diventano rapidamente difettosi dal software stesso.

Ad esempio, YouTube ha 1,9 miliardi di utenti, che guardano oltre 1 miliardo di ore di video YouTube al giorno. Il suo algoritmo di raccomandazione (creato da Google), progettato per ottimizzare il tempo di visualizzazione, è responsabile di circa il 70% dei contenuti guardati. Ma c'era un problema: ciò ha portato a cicli di feedback fuori controllo, portando il *New York Times* a pubblicare il titolo ["YouTube ha scatenato un boom di teorie della cospirazione. Può essere contenuto?"](https://www.nytimes.com/2019/02/19/technology/youtube-conspiracy-stars.html). Apparentemente i sistemi di raccomandazione prevedono quali contenuti piaceranno alle persone, ma hanno anche molto potere nel determinare quali contenuti le persone vedranno.

### Pregiudizio: la professoressa Latanya Sweeney "arrestata"

La dottoressa Latanya Sweeney è professoressa ad Harvard e direttrice del laboratorio sulla privacy dei dati dell'università. Nell'articolo ["Discrimination in Online Ad Delivery"](https://arxiv.org/abs/1301.6822) (vedi <<latanya_arrested>>) descrive la sua scoperta che cercando su Google il suo nome risultavano annunci pubblicitari che dicevano "Latanya Sweeney, arrestata?" anche se è l'unica Latanya Sweeney conosciuta e non è mai stata arrestata. Tuttavia, quando ha cercato su Google altri nomi, come "Kirsten Lindquist", ha ottenuto annunci più neutrali, anche se Kirsten Lindquist è stata arrestata tre volte.

## Integrazione del machine learning con la progettazione del prodotto

Presumibilmente il motivo per cui stai facendo questo lavoro è perché speri che venga utilizzato per qualcosa. Altrimenti stai solo perdendo tempo. Quindi, partiamo dal presupposto che il tuo lavoro finirà da qualche parte. Ora, mentre raccogli i tuoi dati e sviluppi il tuo modello, stai prendendo molte decisioni. A quale livello di aggregazione memorizzerai i tuoi dati? Quale funzione di perdita dovresti usare? Quali set di validazione e formazione dovresti utilizzare? Queste non sono solo domande sull’algoritmo. Sono domande sulla progettazione del prodotto basato sui dati. Ma i product manager, i dirigenti, i giudici, i giornalisti, i medici… chiunque finisca per sviluppare e utilizzare il sistema di cui il tuo modello fa parte non sarà nella posizione ideale per comprendere le decisioni che hai preso, per non parlare di cambiarle.

## Argomenti di etica dei dati

L’etica dei dati è un campo vasto e non possiamo coprire tutto. Sceglieremo invece alcuni argomenti che riteniamo particolarmente rilevanti:

- La necessità di ricorso e responsabilità
- Circuiti di feedback
- Pregiudizio
- Disinformazione

Diamo un'occhiata a ciascuno di essi a turno.

### Ricorso e responsabilità

In un sistema complesso, è facile che nessuno si senta responsabile dei risultati. Sebbene ciò sia comprensibile, non porta a buoni risultati. Nell’esempio precedente del sistema sanitario dell’Arkansas in cui un bug ha portato le persone affette da paralisi cerebrale a perdere l’accesso alle cure necessarie, il creatore dell’algoritmo ha incolpato i funzionari governativi, e i funzionari governativi hanno incolpato coloro che hanno implementato il software. La professoressa della New York University [Danah Boyd](https://www.youtube.com/watch?v=NTl0yyPqf3E) ha descritto questo fenomeno: "La burocrazia è stata spesso utilizzata per spostare o eludere la responsabilità... I sistemi algoritmici di oggi stanno estendendo la burocrazia".

Un ulteriore motivo per cui il ricorso è così necessario è perché i dati spesso contengono errori. I meccanismi di audit e di correzione degli errori sono cruciali. Un database di sospetti membri di bande gestito dalle forze dell'ordine della California è risultato pieno di errori, inclusi 42 bambini che erano stati aggiunti al database quando avevano meno di 1 anno (28 dei quali erano contrassegnati come "che ammettevano di essere membri di una banda"). In questo caso, non esisteva alcun processo per correggere gli errori o rimuovere le persone una volta aggiunte. Un altro esempio è il sistema dei rapporti di credito statunitense: in uno studio su larga scala condotto dalla Federal Trade Commission (FTC) nel 2012, è emerso che il 26% dei consumatori aveva almeno un errore nei propri file e il 5% aveva errori che potevano essere devastanti.  Tuttavia, il processo per correggere tali errori è incredibilmente lento e opaco. Quando il giornalista della radio pubblica [Bobby Allyn](https://www.washingtonpost.com/posteverything/wp/2016/09/08/how-the-careless-errors-of-credit-reporting-agencies-are-ruining-peoples-lives/) scoprì di essere stato erroneamente indicato come colpevole di armi da fuoco, gli ci vollero "più di una dozzina di telefonate, l'opera di un impiegato del tribunale di contea e sei settimane per risolvere il problema. E questo solo dopo aver contattato il dipartimento di comunicazione dell'azienda come giornalista."

Come professionisti del machine learning, non sempre consideriamo nostra responsabilità capire come i nostri algoritmi finiscono per essere implementati nella pratica. Ma dobbiamo farlo.

### Circuiti di feedback

Abbiamo spiegato nel <<capitolo_intro>> come un algoritmo può interagire con il suo ambiente per creare un ciclo di feedback, facendo previsioni che rafforzano le azioni intraprese nel mondo reale, che portano a previsioni ancora più pronunciate nella stessa direzione. 
Ad esempio, consideriamo ancora una volta il sistema di consigli di YouTube. Un paio di anni fa il team di Google ha parlato di come avevano introdotto l'apprendimento per rinforzo (strettamente correlato al deep learning, ma in cui la funzione di perdita rappresenta un risultato potenzialmente molto tempo dopo che si verifica un'azione) per migliorare il sistema di consigli di YouTube. Hanno descritto come hanno utilizzato un algoritmo che forniva raccomandazioni tali da ottimizzare il tempo di visualizzazione.

Tuttavia, gli esseri umani tendono ad essere attratti da contenuti controversi. Ciò significava che i video su argomenti come le teorie del complotto hanno iniziato a essere consigliati sempre di più dal sistema di raccomandazione. Inoltre, si scopre che il tipo di persone interessate alle teorie del complotto sono anche persone che guardano molti video online! Quindi, hanno iniziato ad essere attratti sempre di più da YouTube. Il numero crescente di teorici della cospirazione che guardano video su YouTube ha portato l'algoritmo a raccomandare sempre più teorie del complotto e altri contenuti estremisti, il che ha portato più estremisti a guardare video su YouTube e più persone che guardano YouTube a sviluppare opinioni estremiste, il che ha portato l'algoritmo a raccomandare contenuti più estremisti... Il sistema stava andando fuori controllo.

E questo fenomeno non era contenuto in questo particolare tipo di contenuti. Nel giugno 2019 il *New York Times* ha pubblicato un articolo sul sistema di raccomandazioni di YouTube, intitolato ["On YouTube’s Digital Playground, an Open Gate for Pedophiles"](https://www.nytimes.com/2019/06/03/world/americas/youtube-pedophiles.html). L’articolo iniziava con questa storia agghiacciante:

> : Christiane C. non ci ha pensato niente quando sua figlia di 10 anni e un'amica hanno caricato un video in cui giocavano in una piscina in giardino... Pochi giorni dopo... il video ha avuto migliaia di visualizzazioni. In poco tempo, è arrivato a 400.000... "Ho rivisto il video e mi sono spaventata per il numero di visualizzazioni", ha detto Christiane. Aveva motivo di esserlo. Il sistema di consigli automatizzato di YouTube... aveva iniziato a mostrare il video agli utenti che guardavano altri video di bambini in età prepuberale, parzialmente vestiti, ha scoperto un team di ricercatori.

> : Di per sé, ogni video potrebbe essere perfettamente innocente, un filmino fatto in casa, per esempio, realizzato da un bambino. Eventuali fotogrammi rivelatori sono fugaci e appaiono accidentali. Ma, raggruppati insieme, i loro tratti comuni diventano inconfondibili.

L'algoritmo di raccomandazione di YouTube aveva iniziato a curare playlist per pedofili, selezionando home video innocenti che contenevano bambini in età prepuberale e parzialmente vestiti. 

Nessuno in Google aveva intenzione di creare un sistema che trasformasse i video di famiglia in porno per pedofili. Allora cosa è successo?

Parte del problema qui è la centralità dei parametri nel guidare un sistema finanziariamente importante. Quando un algoritmo ha una metrica da ottimizzare, come hai visto, farà tutto il possibile per ottimizzare quel numero. Ciò tende a portare a tutti i tipi di casi limite e gli esseri umani che interagiscono con un sistema cercheranno, troveranno e sfrutteranno questi casi limite e cicli di feedback a proprio vantaggio.

Ci sono segnali che questo è esattamente ciò che è successo con il sistema di raccomandazioni di YouTube. *The Guardian* ha pubblicato un articolo intitolato ["How an ex-YouTube Insider Investigation its Secret Algorithm"](https://www.theguardian.com/technology/2018/feb/02/youtube-algorithm-election-clinton-trump-guillaume-chaslot) su Guillaume Chaslot, un ex ingegnere di YouTube che ha creato AlgoTransparency, che tiene traccia di questi problemi. Chaslot ha pubblicato il grafico su <<ethics_yt_rt>>, in seguito alla pubblicazione del "Rapporto sulle indagini sull'interferenza russa nelle elezioni presidenziali del 2016" di Robert Mueller.

<img src="images/ethics/image18.jpeg" id="ethics_yt_rt" caption="Copertura del rapporto Mueller" alt="Copertura del rapporto Mueller" width="500">

La copertura del rapporto Mueller da parte di Russia Today è stata un'anomalia estrema in termini di numero di canali che lo raccomandavano. Ciò suggerisce la possibilità che Russia Today, un media russo di proprietà statale, sia riuscito a ingannare l'algoritmo di raccomandazione di YouTube. Sfortunatamente, la mancanza di trasparenza di sistemi come questo rende difficile scoprire il tipo di problemi di cui stiamo discutendo.

Uno dei nostri revisori di questo libro, Aurélien Géron, ha guidato il team di classificazione dei video di YouTube dal 2013 al 2016 (ben prima degli eventi discussi qui). Ha sottolineato che non sono solo i circuiti di feedback che coinvolgono gli esseri umani a costituire un problema. Possono esserci cicli di feedback anche senza esseri umani! Ci ha raccontato un esempio tratto da YouTube:

> : Un segnale importante per classificare l'argomento principale di un video è il canale da cui proviene. Ad esempio, è molto probabile che un video caricato su un canale di cucina sia un video di cucina. Ma come facciamo a sapere di quale argomento tratta un canale? Beh… anche guardando gli argomenti dei video che contiene! Vedi il cappio? Ad esempio, molti video hanno una descrizione che indica quale fotocamera è stata utilizzata per girare il video. Di conseguenza, alcuni di questi video potrebbero essere classificati come video sulla “fotografia”. Se un canale ha un video così classificato erroneamente, potrebbe essere classificato come canale di "fotografia", rendendo ancora più probabile che i futuri video su questo canale vengano erroneamente classificati come "fotografia". Ciò potrebbe persino portare a classificazioni simili a quelle dei virus! Un modo per interrompere questo ciclo di feedback è classificare i video con e senza il segnale del canale. Quindi quando si classificano i canali è possibile utilizzare solo le classi ottenute senza il segnale del canale. In questo modo il ciclo di feedback viene interrotto.

Ci sono esempi positivi di persone e organizzazioni che tentano di combattere questi problemi. Evan Estola, ingegnere capo del machine learning presso Meetup, [ha discusso l'esempio](https://www.youtube.com/watch?v=MqoRzNhrTnQ) di uomini che esprimono più interesse delle donne nei meetup tecnologici. prendere in considerazione il genere potrebbe quindi far sì che l’algoritmo di Meetup raccomandi meno incontri tecnologici alle donne e, di conseguenza, meno donne scoprirebbero e parteciperebbero a incontri tecnologici, il che potrebbe far sì che l’algoritmo suggerisca ancora meno incontri tecnologici alle donne, e così via in un ciclo di feedback auto-rinforzante. Quindi, Evan e il suo team hanno preso la decisione etica di non creare un tale ciclo di feedback per il loro algoritmo di raccomandazione, non utilizzando esplicitamente il genere per quella parte del loro modello. È incoraggiante vedere un’azienda non solo ottimizzare sconsideratamente un parametro, ma considerarne l’impatto. Secondo Evan, "Devi decidere quale funzionalità non utilizzare nel tuo algoritmo... l'algoritmo più ottimale forse non è quello migliore da lanciare in produzione".

Mentre Meetup ha scelto di evitare un simile risultato, Facebook fornisce un esempio di come si lascia scatenare un ciclo di feedback incontrollato. Come YouTube, tende a radicalizzare gli utenti interessati a una teoria del complotto introducendoli ad altre. Come Renee DiResta, ricercatrice sulla proliferazione della disinformazione, [scrive](https://www.fastcompany.com/3059742/social-network-algorithms-are-distorting-reality-by-boosting-conspiracy-theories):

> : Una volta che le persone si uniscono a un singolo gruppo [Facebook] con una mentalità complottista, vengono indirizzate algoritmicamente a una pletora di altri. Unisciti a un gruppo anti-vaccini e i tuoi suggerimenti includeranno anti-OGM, sorveglianza sulle scie chimiche, Terrapiattista (sì, davvero) e "gruppi che curano il cancro in modo naturale. Invece di tirare fuori un utente dalla tana del coniglio, il motore di raccomandazione lo spinge ulteriormente dentro."

È estremamente importante tenere presente che questo tipo di comportamento può verificarsi e anticipare un ciclo di feedback o intraprendere azioni positive per interromperlo quando ne vedi i primi segni nei tuoi progetti. Un'altra cosa da tenere a mente è il *bias* che, come abbiamo discusso brevemente nel capitolo precedente, può interagire con i cicli di feedback in modi molto problematici.

### Pregiudizio

Le discussioni sui pregiudizi online tendono a diventare piuttosto confuse abbastanza velocemente. La parola "bias" significa tante cose diverse. Gli statistici spesso pensano che quando gli esperti di etica dei dati parlano di bias si riferiscano alla definizione statistica del termine bias. Ma non lo sono. E certamente non stanno parlando dei pregiudizi che compaiono nei pesi e nei pregiudizi che sono i parametri del tuo modello!

Ciò di cui stanno parlando è il concetto di pregiudizio nelle scienze sociali. In ["A Framework for Understanding Unintended Consequences of Machine Learning"](https://arxiv.org/abs/1901.10002) Harini Suresh e John Guttag del MIT descrivono sei tipi di bias nell'apprendimento automatico, riassunti in <<bias>> dal loro articolo.

Guttag)" alt="Un diagramma che mostra tutte le fonti in cui i pregiudizi possono apparire nell'apprendimento automatico" width="700">

Discuteremo quattro di questi tipi di pregiudizi, quelli che abbiamo trovato più utili nel nostro lavoro (vedi il documento per i dettagli sugli altri).

#### Pregiudizio storico

Il *pregiudizio storico* deriva dal fatto che le persone sono prevenute, i processi sono prevenuti e la società è prevenuta. Suresh e Guttag affermano: "Il pregiudizio storico è un problema fondamentale e strutturale con la prima fase del processo di generazione dei dati e può esistere anche con un campionamento e una selezione delle caratteristiche perfetti".

Ad esempio, ecco alcuni esempi di *pregiudizi razziali* storici negli Stati Uniti, tratti dall'articolo del *New York Times* ["Racial Bias, Even When We Have Good Intentions"](https://www.nytimes.com/2015/01/04/upshot/the-measuring-sticks-of-racial-bias-.html) di Sendhil Mullainathan dell'Università di Chicago:

  - Quando ai medici venivano mostrati file identici, erano molto meno propensi a raccomandare il cateterismo cardiaco (una procedura utile) ai pazienti neri.
  - Durante la contrattazione per un'auto usata, ai neri veniva offerto un prezzo iniziale di 700 dollari più alto e ricevevano concessioni molto più piccole.
  - Rispondere agli annunci di appartamenti in affitto su Craigslist con un nome nero ha suscitato meno risposte rispetto a un nome bianco.
  - Una giuria composta esclusivamente da bianchi aveva 16 punti percentuali in più di probabilità di condannare un imputato nero rispetto a uno bianco, ma quando una giuria aveva un membro nero condannava entrambi allo stesso tasso.

L'algoritmo COMPAS, ampiamente utilizzato per le sentenze e le decisioni su cauzione negli Stati Uniti, è un esempio di un importante algoritmo che, quando testato da [ProPublica](https://www.propublica.org/article/machine-bias-risk-assessments-in-criminal-sentencing), ha mostrato chiari pregiudizi razziali nella pratica (<<bail_algorithm>>).

<img src="images/ethics/image6.png" id="bail_algorithm" caption="Risultati dell'algoritmo COMPAS" alt="La tabella che mostra che l'algoritmo COMPAS ha maggiori probabilità di concedere la cauzione ai bianchi, anche se recidivano di più" width="700">

Qualsiasi set di dati che coinvolga esseri umani può avere questo tipo di pregiudizio: dati medici, dati sulle vendite, dati sugli alloggi, dati politici e così via. Poiché la distorsione di fondo è così pervasiva, la distorsione nei set di dati è molto pervasiva. I pregiudizi razziali si manifestano anche nella visione artificiale, come mostrato nell'esempio delle foto categorizzate automaticamente condivise su Twitter da un utente di Google Foto mostrate in <<google_photos>>.

<img src="images/ethics/image7.png" id="google_photos" caption="Una di queste etichette è molto sbagliata..." alt="Screenshot dell'uso di Google Foto che etichettano un utente nero e la sua amica come gorilla" width="450">

Sì, questo dimostra quello che pensi che sia: Google Foto ha classificato la foto di un utente nero con il suo amico come "gorilla"! Questo passo falso algoritmico ha attirato molta attenzione da parte dei media. "Siamo sconvolti e sinceramente dispiaciuti che ciò sia accaduto", ha detto un portavoce dell'azienda. “C’è chiaramente ancora molto lavoro da fare con l’etichettatura automatica delle immagini e stiamo cercando come evitare che questo tipo di errori si ripetano in futuro”.

Sfortunatamente, risolvere i problemi nei sistemi di apprendimento automatico quando i dati di input presentano problemi è difficile. Il primo tentativo di Google non ha ispirato fiducia, come ha suggerito il servizio *The Guardian* (<<gorilla-ban>>).

<img src="images/ethics/image8.png" id="gorilla-ban" caption="La prima risposta di Google al problema" alt="Immagini di un titolo del Guardian, che mostra che Google ha rimosso gorilla e altri soldi dalle possibili etichette del suo algoritmo" width="500">

Questi tipi di problemi non si limitano certamente solo a Google. I ricercatori del MIT hanno studiato le API di visione artificiale online più popolari per vedere quanto fossero precise. Ma non si sono limitati a calcolare un singolo numero di precisione: hanno invece esaminato la precisione in quattro gruppi diversi, come illustrato in <<face_recognition>>.

<img src="images/ethics/image9.jpeg" id="face_recognition" caption="Tasso di errore per sesso e razza per vari sistemi di riconoscimento facciale" alt="Tabella che mostra come i vari sistemi di riconoscimento facciale funzionano molto peggio sulle tonalità più scure della pelle e sulle donne" width="600">

Il sistema IBM, ad esempio, aveva un tasso di errore del 34,7% per le donne più scure, contro lo 0,3% per i maschi più chiari: oltre 100 volte più errori! Alcune persone hanno reagito erroneamente a questi esperimenti sostenendo che la differenza era semplicemente dovuta al fatto che la pelle più scura è più difficile da riconoscere per i computer. Tuttavia, ciò che in realtà è successo è che, dopo la pubblicità negativa creata da questo risultato, tutte le aziende in questione hanno migliorato notevolmente i loro modelli per la pelle più scura, tanto che un anno dopo erano quasi altrettanto validi che per la pelle più chiara. Ciò che questo in realtà ha dimostrato è che gli sviluppatori non sono riusciti a utilizzare set di dati contenenti un numero sufficiente di volti più scuri o a testare il loro prodotto con volti più scuri.

Una delle ricercatrici del MIT, Joy Buolamwini, ha avvertito: "Siamo entrati nell'era dell'automazione troppo sicuri di sé ma impreparati. Se non riusciamo a creare un'intelligenza artificiale etica e inclusiva, rischiamo di perdere i progressi ottenuti nei diritti civili e nell'uguaglianza di genere con il pretesto della neutralità della macchina."

Parte del problema sembra essere uno squilibrio sistematico nella composizione dei set di dati più diffusi utilizzati per l’addestramento dei modelli. L'abstract dell'articolo ["No Classification Without Representation: Assessing Geodiversity Issues in Open Data Sets for the Developing World"](https://arxiv.org/abs/1711.08536) di Shreya Shankar et al. afferma: "Analizziamo due grandi set di dati di immagini disponibili al pubblico per valutare la geo-diversità e scopriamo che questi set di dati sembrano mostrare un osservabile pregiudizio di rappresentazione amerocentrico ed eurocentrico. Inoltre, analizziamo i classificatori addestrati su questi set di dati per valutare l'impatto di queste distribuzioni di addestramento e trovare forti differenze nelle prestazioni relative su immagini provenienti da luoghi diversi. " <<image_provenance>> mostra uno dei grafici dell'articolo, che mostra la composizione geografica di quelli che erano, all'epoca (e lo sono ancora, mentre questo libro viene scritto) i due set di dati di immagini più importanti per i modelli di addestramento.

La stragrande maggioranza delle immagini proviene dagli Stati Uniti e da altri paesi occidentali, il che porta i modelli addestrati su ImageNet a ottenere risultati peggiori su scene di altri paesi e culture. Ad esempio, la ricerca ha scoperto che tali modelli sono peggiori nell’identificare gli articoli domestici (come sapone, spezie, divani o letti) provenienti da paesi a basso reddito. <<object_detect>> mostra un'immagine dal documento, ["Il riconoscimento degli oggetti funziona per tutti?"](https://arxiv.org/pdf/1906.02659.pdf) di Terrance DeVries et al. di Facebook AI Research che illustra questo punto.

In questo esempio, possiamo vedere che l'esempio del sapone a basso reddito è molto lontano dall'essere accurato, poiché ogni servizio di riconoscimento di immagini commerciali prevede "cibo" come la risposta più probabile!

Come discuteremo tra poco, inoltre, la stragrande maggioranza dei ricercatori e sviluppatori di intelligenza artificiale sono giovani uomini bianchi. La maggior parte dei progetti che abbiamo visto eseguono la maggior parte dei test sugli utenti utilizzando amici e familiari del gruppo di sviluppo del prodotto più vicino. Detto questo, il tipo di problemi di cui abbiamo appena discusso non dovrebbe sorprendere.

Un simile pregiudizio storico si trova nei testi utilizzati come dati per i modelli di elaborazione del linguaggio naturale. Ciò emerge in molti modi nelle attività di machine learning a valle. Ad esempio, [è stato ampiamente riportato](https://nypost.com/2017/11/30/google-translates-algorithm-has-a-gender-bias/) che fino all'anno scorso Google Translate mostrava un pregiudizio sistematico nel modo in cui traduceva il pronome turco neutro rispetto al genere "o" in inglese: quando applicato a lavori che sono spesso associati agli uomini usava "lui", e quando applicato a lavori che sono spesso associati alle donne usava "lei" (<<turco_genere>>).

Vediamo questo tipo di pregiudizio anche nelle pubblicità online. Ad esempio, uno [studio](https://arxiv.org/abs/1904.02095) nel 2019 di Muhammad Ali et al. ha scoperto che anche quando la persona che inserisce l’annuncio non discrimina intenzionalmente, Facebook mostrerà gli annunci a un pubblico molto diverso in base alla razza e al sesso. Annunci immobiliari con lo stesso testo, ma che raffiguravano una famiglia bianca o nera, venivano mostrati a un pubblico razzialmente diverso.

#### Distorsione nella misurazione

Nell'articolo ["Does Machine Learning Automate Moral Hazard and Error"](https://scholar.harvard.edu/files/sendhil/files/aer.p20171084.pdf) in *American Economic Review*, Sendhil Mullainathan e Ziad Obermeyer esaminano un modello che cerca di rispondere alla domanda: utilizzando i dati storici delle cartelle cliniche elettroniche (EHR), quali fattori sono maggiormente predittivi dell'ictus? Questi sono i principali predittori del modello:

  - Precedente ictus
  - Malattie cardiovascolari
  - Lesioni accidentali
  - Nodulo al seno benigno
  - Colonscopia
  - Sinusite

Tuttavia, solo i primi due hanno qualcosa a che fare con un ictus! Sulla base di ciò che abbiamo studiato finora, probabilmente puoi indovinare il perché. Non abbiamo realmente misurato l’*ictus*, che si verifica quando a una regione del cervello viene negato l’ossigeno a causa di un’interruzione dell’afflusso di sangue. Ciò che abbiamo misurato è chi ha avuto sintomi, è andato da un medico, ha effettuato i test appropriati e *e* ha ricevuto una diagnosi di ictus. In realtà avere un ictus non è l'unica cosa correlata a questo elenco completo: è anche correlato all'essere il tipo di persona che va effettivamente dal medico (che è influenzato da chi ha accesso all'assistenza sanitaria, può permettersi il ticket, non subisce discriminazioni mediche basate sulla razza o sul genere e altro ancora)! Se è probabile che tu vada dal medico per un *infortunio accidentale*, è ​​probabile che andrai dal medico anche quando hai un ictus.

Questo è un esempio di *bias di misurazione*. Si verifica quando i nostri modelli commettono errori perché stiamo misurando la cosa sbagliata, o nel modo sbagliato, o incorporiamo quella misura nel modello in modo inappropriato.

#### Distorsione da aggregazione

Il *bias di aggregazione* si verifica quando i modelli non aggregano i dati in modo da incorporare tutti i fattori appropriati o quando un modello non include i termini di interazione necessari, le non linearità e così via. Ciò può verificarsi soprattutto in ambito medico. Ad esempio, il modo in cui viene trattato il diabete si basa spesso su semplici statistiche univariate e studi che coinvolgono piccoli gruppi di persone eterogenee. L’analisi dei risultati viene spesso effettuata in modo da non tenere conto delle diverse etnie o dei generi. Tuttavia, risulta che i pazienti diabetici hanno [complicanze diverse a seconda delle etnie] (https://www.ncbi.nlm.nih.gov/pubmed/24037313) e i livelli di HbA1c (ampiamente utilizzati per diagnosticare e monitorare il diabete) [differiscono in modo complesso a seconda delle etnie e dei generi] (https://www.ncbi.nlm.nih.gov/pubmed/22238408). Ciò può comportare che le persone vengano diagnosticate erroneamente o trattate in modo errato perché le decisioni mediche si basano su un modello che non include queste importanti variabili e interazioni.

#### Distorsioni di rappresentazione

L'abstract dell'articolo ["Bias in Bios: A Case Study of Semantic Representation Bias in a High-Stakes setting"](https://arxiv.org/abs/1901.09451) di Maria De-Arteaga et al. rileva che esiste uno squilibrio di genere nelle occupazioni (ad esempio, le donne hanno maggiori probabilità di essere infermiere e i maschi hanno maggiori probabilità di essere pastori) e afferma che: "le differenze nei tassi positivi reali tra i sessi sono correlate con gli squilibri di genere esistenti nelle occupazioni, che possono aggravare questi squilibri".

In altre parole, i ricercatori hanno notato che i modelli di previsione dell’occupazione non solo *riflettono* l’effettivo squilibrio di genere nella popolazione sottostante, ma in realtà lo *amplificano*! Questo tipo di *distorsione di rappresentazione* è abbastanza comune, in particolare per i modelli semplici. Quando esiste una relazione sottostante chiara e facile da vedere, un modello semplice spesso presuppone semplicemente che questa relazione sia sempre valida. Come mostra <<representation_bias>> dal documento, per le occupazioni che avevano una percentuale più elevata di donne, il modello tendeva a sovrastimare la prevalenza di tale occupazione.

<img src="images/ethics/image12.png" id="representation_bias" caption="Errore del modello nella previsione dell'occupazione rispetto alla percentuale di donne in detta occupazione" alt="Grafico che mostra come le previsioni del modello amplificano eccessivamente i pregiudizi esistenti" width="500">

Ad esempio, nel set di dati di formazione il 14,6% dei chirurghi erano donne, mentre nelle previsioni del modello solo l’11,6% dei veri positivi erano donne. Il modello amplifica quindi la distorsione esistente nel training set.

Ora che abbiamo visto che questi pregiudizi esistono, cosa possiamo fare per mitigarli?

### Affrontare diversi tipi di pregiudizi

Diversi tipi di bias richiedono approcci diversi per la mitigazione. Sebbene la raccolta di un set di dati più diversificato possa risolvere i bias di rappresentazione, ciò non aiuterebbe con i bias storici o di misurazione.  Tutti i set di dati contengono bias.  Non esiste un set di dati completamente distorto.  Molti ricercatori nel settore stanno convergendo su una serie di proposte per consentire una migliore documentazione delle decisioni, del contesto e delle specifiche su come e perché è stato creato un particolare set di dati, in quali scenari è appropriato utilizzarlo e quali sono le limitazioni. In questo modo, coloro che utilizzano un particolare set di dati non saranno colti di sorpresa dai suoi pregiudizi e limiti.

Spesso sentiamo la domanda: "Gli esseri umani sono di parte, quindi i pregiudizi algoritmici contano?" Questo accade così spesso che ci deve essere qualche ragionamento che abbia senso per le persone che lo chiedono, ma a noi non sembra molto logico! Indipendentemente dal fatto che ciò sia logicamente valido, è importante rendersi conto che gli algoritmi (in particolare gli algoritmi di apprendimento automatico!) e le persone sono diversi. Considera questi punti sugli algoritmi di machine learning:

  - _L'apprendimento automatico può creare cicli di feedback_: Piccole quantità di bias possono aumentare rapidamente in modo esponenziale a causa dei cicli di feedback.
  - _L'apprendimento automatico può amplificare i pregiudizi_: i pregiudizi umani possono portare a quantità maggiori di pregiudizi legati all'apprendimento automatico.
  - _Algoritmi e esseri umani vengono utilizzati in modo diverso_: nella pratica i decisori umani e i decisori algoritmici non vengono utilizzati in modo intercambiabile plug-and-play.
  - _La tecnologia è potere_:: E con ciò deriva la responsabilità.

Come ha dimostrato l’esempio del settore sanitario dell’Arkansas, l’apprendimento automatico viene spesso implementato nella pratica non perché porti a risultati migliori, ma perché è più economico ed efficiente. Cathy O'Neill, nel suo libro *Weapons of Math Destruction* (Crown), ha descritto il modello di come i privilegiati vengono elaborati dalle persone, mentre i poveri vengono elaborati dagli algoritmi. Questo è solo uno dei tanti modi in cui gli algoritmi vengono utilizzati in modo diverso rispetto ai decisori umani. Altri includono:

  - Le persone sono più propense a ritenere che gli algoritmi siano oggettivi o privi di errori (anche se viene data loro la possibilità di un intervento umano).
  - È più probabile che gli algoritmi vengano implementati senza alcuna procedura di ricorso in atto.
  - Gli algoritmi sono spesso utilizzati su larga scala.
  - I sistemi algoritmici sono economici.

Anche in assenza di pregiudizi, gli algoritmi (e in particolare il deep learning, poiché è un algoritmo così efficace e scalabile) possono portare a problemi sociali negativi, come quando vengono utilizzati per la *disinformazione*.

### Disinformazione

La *disinformazione* ha una storia che risale a centinaia o addirittura migliaia di anni fa. Non si tratta necessariamente di indurre qualcuno a credere a qualcosa di falso, ma piuttosto spesso viene utilizzato per seminare disarmonia e incertezza e per convincere le persone a rinunciare alla ricerca della verità.  Ricevere account in conflitto può indurre le persone a presumere che non potranno mai sapere di chi o di cosa fidarsi.

Alcune persone pensano che la disinformazione riguardi principalmente informazioni false o *notizie false*, ma in realtà la disinformazione può spesso contenere semi di verità o mezze verità estrapolate dal contesto. Ladislav Bittman era un ufficiale dell'intelligence dell'URSS che in seguito disertò negli Stati Uniti e scrisse alcuni libri negli anni '70 e '80 sul ruolo della disinformazione nelle operazioni di propaganda sovietica. In *Il KGB e la disinformazione sovietica* (Pergamo) scrisse: "La maggior parte delle campagne sono una miscela attentamente progettata di fatti, mezze verità, esagerazioni e bugie deliberate".

Negli Stati Uniti questo ha colpito da vicino negli ultimi anni, con l’FBI che ha descritto nei dettagli una massiccia campagna di disinformazione collegata alla Russia nelle elezioni del 2016. Comprendere la disinformazione utilizzata in questa campagna è molto istruttivo. Ad esempio, l’FBI ha scoperto che la campagna di disinformazione russa spesso organizzava due false proteste “di base”, una per ciascuna parte di una questione, e le spingeva a protestare allo stesso tempo! Il [*Houston Chronicle*](https://www.houstonchronicle.com/local/gray-matters/article/A-Houston-protest-organized-by-Russian-trolls-12625481.php) ha riportato uno di questi strani eventi (<<texas>>).

> : Un gruppo che si autodefiniva "Cuore del Texas" l'aveva organizzato sui social media: una protesta, dicevano, contro l'"islamizzazione" del Texas. Su un lato di Travis Street ho trovato circa 10 manifestanti. Dall’altro lato ho trovato circa 50 contromanifestanti. Ma non sono riuscito a trovare gli organizzatori del raduno. Nessun "cuore del Texas". Ho pensato che fosse strano e l'ho menzionato nell'articolo: Che tipo di gruppo non si presenta al proprio evento? Ora so perché. A quanto pare, in quel momento gli organizzatori del raduno si trovavano a San Pietroburgo, in Russia. "Heart of Texas" è uno dei gruppi di troll internet citati nella recente accusa del procuratore speciale Robert Mueller contro i russi che tentano di manomettere le elezioni presidenziali americane.

<img src="images/ethics/image13.png" id="texas" caption="Evento organizzato dal gruppo Heart of Texas" alt="Screenshot di un evento organizzato dal gruppo Heart of Texas" width="300">

La disinformazione spesso implica campagne coordinate di comportamenti non autentici.  Ad esempio, gli account fraudolenti potrebbero provare a far sembrare che molte persone abbiano un punto di vista particolare.  Anche se alla maggior parte di noi piace pensare a noi stessi come persone dalla mentalità indipendente, in realtà ci siamo evoluti per essere influenzati dagli altri nel nostro gruppo e in opposizione a quelli del nostro gruppo esterno.  Le discussioni online possono influenzare i nostri punti di vista o alterare la gamma di ciò che consideriamo punti di vista accettabili. Gli esseri umani sono animali sociali e come animali sociali siamo estremamente influenzati dalle persone che ci circondano. La radicalizzazione avviene sempre più spesso negli ambienti online; l'influenza proviene dalle persone nello spazio virtuale dei forum online e dei social network.

La disinformazione attraverso il testo generato automaticamente è un problema particolarmente significativo, a causa della capacità notevolmente aumentata fornita dal deep learning. Discutiamo questo argomento in modo approfondito quando approfondiremo la creazione di modelli linguistici, in <<chapter_nlp>>.

Un approccio proposto è quello di sviluppare una qualche forma di firma digitale, implementarla in modo fluido e creare norme secondo le quali dovremmo fidarci solo del contenuto che è stato verificato. Il capo dell'Allen Institute on AI, Oren Etzioni, ha scritto una proposta del genere in un articolo intitolato ["How Will We Prevent AI-Based Forgery?"](https://hbr.org/2019/03/how-will-we-prevent-ai-based-forgery): "L'intelligenza artificiale è pronta a rendere la falsificazione ad alta fedeltà economica e automatizzata, portando a conseguenze potenzialmente disastrose per la democrazia, la sicurezza e la società. Lo spettro della falsificazione dell'IA significa che abbiamo bisogno agire per rendere le firme digitali di rigore come mezzo di autenticazione dei contenuti digitali."

Anche se non possiamo sperare di discutere tutte le questioni etiche sollevate dal deep learning e dagli algoritmi più in generale, speriamo che questa breve introduzione sia stata un utile punto di partenza su cui costruire. Passeremo ora alle domande su come identificare le questioni etiche e cosa fare al riguardo.

## Identificare e affrontare le questioni etiche

Gli errori accadono. Scoprirli e affrontarli deve far parte della progettazione di qualsiasi sistema che includa l’apprendimento automatico (e anche molti altri sistemi).  Le questioni sollevate nell’ambito dell’etica dei dati sono spesso complesse e interdisciplinari, ma è fondamentale lavorare per affrontarle.

Allora cosa possiamo fare?  Questo è un argomento importante, ma alcuni passi per affrontare le questioni etiche sono:

- Analizza un progetto su cui stai lavorando.
- Implementare processi nella vostra azienda per individuare e affrontare i rischi etici.
- Sostenere una buona politica.
- Aumentare la diversità.

Esaminiamo ciascuno di questi passaggi, iniziando con l'analisi del progetto su cui stai lavorando.

### Analizza un progetto su cui stai lavorando

È facile trascurare questioni importanti quando si considerano le implicazioni etiche del proprio lavoro. Una cosa che aiuta enormemente è semplicemente porre le domande giuste. Rachel Thomas consiglia di considerare le seguenti domande durante lo sviluppo di un progetto dati:

  - Dovremmo farlo?
  - Che distorsione c'è nei dati?
  - È possibile verificare il codice e i dati?
  - Quali sono i tassi di errore per i diversi sottogruppi?
  - Qual è l'accuratezza di una semplice alternativa basata su regole?
  - Quali processi sono in atto per gestire ricorsi o errori?
  - Quanto è diversificato il team che lo ha costruito?

Queste domande potrebbero aiutarti a identificare i problemi in sospeso e le possibili alternative più facili da comprendere e controllare. Oltre a porre le domande giuste, è anche importante considerare le pratiche e i processi da implementare.

Una cosa da considerare in questa fase è quali dati stai raccogliendo e archiviando. I dati spesso finiscono per essere utilizzati per scopi diversi da quelli per cui sono stati originariamente raccolti. Ad esempio, IBM iniziò a vendere alla Germania nazista ben prima dell’Olocausto, aiutando anche il censimento tedesco del 1933 condotto da Adolf Hitler, che fu efficace nell’identificare molti più ebrei di quanto precedentemente riconosciuto in Germania. Allo stesso modo, i dati del censimento statunitense furono utilizzati per radunare i giapponesi-americani (che erano cittadini statunitensi) per l'internamento durante la seconda guerra mondiale. È importante riconoscere come i dati e le immagini raccolti possano essere successivamente utilizzati come armi. Il professore della Columbia [Tim Wu ha scritto](https://www.nytimes.com/2019/04/10/opinion/sunday/privacy-capitalism.html) che "Devi presumere che tutti i dati personali conservati da Facebook o Android siano dati che i governi di tutto il mondo cercheranno di ottenere o che i ladri cercheranno di rubare".

### Processi da implementare

Il Markkula Center ha pubblicato [An Ethical Toolkit for Engineering/Design Practice](https://www.scu.edu/ethics-in-technology-practice/ethical-toolkit/) che include alcune pratiche concrete da implementare presso la vostra azienda, tra cui controlli regolarmente programmati per la ricerca proattiva di rischi etici (in modo simile ai test di penetrazione della sicurezza informatica), espandendo il circolo etico per includere le prospettive di una varietà di parti interessate e considerando le persone terribili (come potrebbero i cattivi attori abusare, rubare, interpretare male, hackerare, distruggere o utilizzare come arma ciò che stai costruendo?). 

Anche se non disponi di un team eterogeneo, puoi comunque provare a includere in modo proattivo le prospettive di un gruppo più ampio, considerando domande come queste (fornite dal Markkula Center):

  - Quali interessi, desideri, competenze, esperienze e valori abbiamo semplicemente assunto, anziché effettivamente consultato?
  - Chi sono tutte le parti interessate che saranno direttamente interessate dal nostro prodotto? Come sono stati tutelati i loro interessi? Come facciamo a sapere quali sono realmente i loro interessi, abbiamo chiesto?
  - Chi/quali gruppi e individui saranno indirettamente influenzati in modo significativo?
  - Chi potrebbe utilizzare questo prodotto che non ci aspettavamo o per scopi che inizialmente non avevamo previsto?

#### Lenti etiche

Un'altra risorsa utile del Markkula Center è il suo [Quadro concettuale nella pratica tecnologica e ingegneristica](https://www.scu.edu/ethics-in-technology-practice/ethical-lenses/). Questo considera come diverse lenti etiche fondamentali possano aiutare a identificare questioni concrete e delinea i seguenti approcci e domande chiave:

  - L'approccio basato sui diritti: quale opzione rispetta meglio i diritti di tutti coloro che hanno un interesse?
  - L'approccio della giustizia: quale opzione tratta le persone in modo equo o proporzionato?
  - L'approccio utilitaristico: quale opzione produrrà il massimo bene e causerà il minor danno?
  - L'approccio al bene comune: quale opzione è più utile alla comunità nel suo insieme e non solo ad alcuni membri?
  - L'approccio della virtù: quale opzione mi porta ad agire come il tipo di persona che voglio essere?

I consigli di Markkula includono un'analisi più approfondita di ciascuna di queste prospettive, inclusa la visione di un progetto attraverso le lenti delle sue *conseguenze*:

  - Chi sarà direttamente interessato da questo progetto? Chi sarà influenzato indirettamente?
  - Gli effetti nel loro insieme creeranno probabilmente più benefici che danni e quali tipi di benefici e danni?
  - Stiamo pensando a tutti i tipi rilevanti di danno/beneficio (psicologico, politico, ambientale, morale, cognitivo, emotivo, istituzionale, culturale)?
  - In che modo questo progetto potrebbe influenzare le generazioni future?
  - I rischi di danni derivanti da questo progetto ricadono in modo sproporzionato sui meno potenti della società? I benefici andranno in modo sproporzionato ai benestanti?
  - Abbiamo considerato adeguatamente il "dual-use"?

La lente alternativa a questa è la prospettiva *deontologica*, che si concentra sui concetti base di *giusto* e *sbagliato*:

  - Quali diritti degli altri e doveri verso gli altri dobbiamo rispettare?
  - In che modo questo progetto potrebbe influire sulla dignità e sull'autonomia di ciascuna parte interessata?
  - Quali considerazioni di fiducia e di giustizia sono rilevanti per questo disegno/progetto?
  - Questo progetto implica doveri morali contrastanti verso gli altri o diritti contrastanti delle parti interessate? Come possiamo dare priorità a questi?

Uno dei modi migliori per trovare risposte complete e ponderate a domande come queste è garantire che le persone che pongono le domande siano *diverse*.

### Il potere della diversità

Attualmente, secondo [uno studio di Element AI](https://medium.com/element-ai-research-lab/estimating-the-gender-ratio-of-ai-researchers-around-the-world-81d2b8dbe9c3), meno del 12% dei ricercatori nel campo dell'intelligenza artificiale sono donne. Le statistiche sono altrettanto disastrose quando si tratta di razza ed età. Quando tutti i membri di un team hanno un background simile, è probabile che abbiano punti ciechi simili riguardo ai rischi etici. L'*Harvard Business Review* (HBR) ha pubblicato una serie di studi che mostrano i numerosi vantaggi di team diversi, tra cui:

- ["Come la diversità può guidare l'innovazione"](https://hbr.org/2013/12/how-diversity-can-drive-innovation)
- ["I team risolvono i problemi più velocemente quando sono più cognitivamente diversi"](https://hbr.org/2017/03/teams-solve-problems-faster-when-theyre-more-cognitively-diverse)
- ["Perché i team diversi sono più intelligenti"](https://hbr.org/2016/11/why-diverse-teams-are-smarter) e
- ["Difendi la tua ricerca: cosa rende un team più intelligente? Più donne"](https://hbr.org/2011/06/defend-your-research-what-makes-a-team-smarter-more-women)

La diversità può portare a identificare prima i problemi e a prendere in considerazione una gamma più ampia di soluzioni. Ad esempio, Tracy Chou è stata una delle prime ingegneri di Quora. Lei [ha scritto delle sue esperienze](https://qz.com/1016900/tracy-chou-leading-silicon-valley-engineer-explains-why-every-tech-worker-needs-a-humanities-education/), descrivendo come ha sostenuto internamente l'aggiunta di una funzionalità che consentisse di bloccare i troll e altri cattivi attori. Chou racconta: "Ero ansioso di lavorare sulla funzione perché personalmente mi sentivo antagonizzato e abusato sul sito (il genere non è una ragione improbabile per spiegare il perché)... Ma se non avessi avuto quella prospettiva personale, è possibile che il team di Quora non avrebbe dato priorità alla creazione di un pulsante di blocco così presto nella sua esistenza." Le molestie spesso spingono le persone appartenenti a gruppi emarginati ad abbandonare le piattaforme online, quindi questa funzionalità è stata importante per preservare la salute della community di Quora.

Un aspetto cruciale da comprendere è che, secondo la [*Harvard Business Review*](https://www.researchgate.net/publication/268325574_By_RESEARCH_REPORT_The_Athena_Factor_Reversing_the_Brain_Drain_in_Science_Engineering_and_Technology), le donne abbandonano il settore tecnologico a una velocità doppia rispetto agli uomini (il 41% delle donne che lavorano nel settore tecnologico lascia il settore tecnologico, rispetto al 17% degli uomini). Da un’analisi di oltre 200 libri, white paper e articoli è emerso che il motivo per cui se ne vanno è che “vengono trattati ingiustamente: sono sottopagati, hanno meno probabilità di essere accelerati rispetto ai loro colleghi maschi e non sono in grado di avanzare”. 

Gli studi hanno confermato una serie di fattori che rendono più difficile per le donne avanzare sul posto di lavoro. Le donne ricevono feedback più vaghi e critiche sulla personalità nelle valutazioni delle prestazioni, mentre gli uomini ricevono consigli pratici legati ai risultati aziendali (il che è più utile). Le donne spesso sperimentano di essere escluse da ruoli più creativi e innovativi e di non ricevere incarichi "estensivi" ad alta visibilità utili per ottenere una promozione. Uno studio ha rilevato che le voci degli uomini sono percepite come più persuasive, basate sui fatti e logiche rispetto alle voci delle donne, anche quando leggono copioni identici.

È stato statisticamente dimostrato che ricevere un tutoraggio aiuta gli uomini ad avanzare, ma non le donne. La ragione di ciò è che quando le donne ricevono un tutoraggio, si tratta di consigli su come dovrebbero cambiare e acquisire maggiore conoscenza di sé. Quando gli uomini ricevono un tutoraggio, si tratta di un sostegno pubblico alla loro autorità. Indovina quale è più utile per ottenere una promozione?

Finché le donne qualificate continueranno ad abbandonare la tecnologia, insegnare a più ragazze a programmare non risolverà i problemi di diversità che affliggono il settore. Le iniziative a favore della diversità spesso finiscono per concentrarsi principalmente sulle donne bianche, anche se le donne di colore devono affrontare molte barriere aggiuntive. Nelle [interviste](https://worklifelaw.org/publications/Double-Jeopardy-Report_v6_full_web-sm.pdf) con 60 donne di colore che lavorano nella ricerca STEM, il 100% ha subito discriminazioni.

Il processo di assunzione è particolarmente interrotto nel settore tecnologico. Uno studio indicativo della disfunzione viene da Triplebyte, una società che aiuta a collocare gli ingegneri del software nelle aziende, conducendo un colloquio tecnico standardizzato come parte di questo processo. Hanno un set di dati affascinante: i risultati di come oltre 300 ingegneri hanno superato l'esame, insieme ai risultati di come hanno fatto quegli ingegneri durante il processo di colloquio per una varietà di aziende. La scoperta numero uno della [ricerca di Triplebyte](https://triplebyte.com/blog/who-y-combinator-companies-want) è che "i tipi di programmatori che ogni azienda cerca spesso hanno poco a che fare con ciò di cui l'azienda ha bisogno o fa. Piuttosto, riflettono la cultura aziendale e il background dei fondatori".

Questa è una sfida per coloro che cercano di entrare nel mondo del deep learning, poiché oggi la maggior parte dei gruppi di deep learning delle aziende sono stati fondati da accademici. Questi gruppi tendono a cercare persone "come loro", cioè persone in grado di risolvere problemi di matematica complessi e comprendere un gergo denso. Non sempre sanno come individuare le persone che sono effettivamente brave a risolvere problemi reali utilizzando il deep learning.

Ciò lascia una grande opportunità alle aziende pronte a guardare oltre lo status e il pedigree e concentrarsi sui risultati!

### Equità, responsabilità e trasparenza

La società professionale degli scienziati informatici, l'ACM, organizza una conferenza sull'etica dei dati chiamata Conference on Fairness, Accountability, and Transparency. "Equità, responsabilità e trasparenza", che in passato veniva usato con l'acronimo *FAT* ma ora viene utilizzato con il meno discutibile *FAccT*. Microsoft ha un gruppo focalizzato su "Equità, responsabilità, trasparenza ed etica" (FATE). In questa sezione utilizzeremo "FAccT" per fare riferimento ai concetti di *equità, responsabilità e trasparenza*.

FAccT è un altro obiettivo che potresti trovare utile nel considerare le questioni etiche. Una risorsa utile a questo scopo è il libro online gratuito [*Fairness and Machine Learning: Limitations and Opportunities*](https://fairmlbook.org/) di Solon Barocas, Moritz Hardt e Arvind Narayanan, che "offre una prospettiva sull'apprendimento automatico che tratta l'equità come una preoccupazione centrale piuttosto che come un ripensamento". Avverte anche, tuttavia, che "ha una portata intenzionalmente ristretta... Una cornice ristretta dell'etica dell'apprendimento automatico potrebbe essere una tentazione per i tecnologi e le imprese come un modo per concentrarsi su interventi tecnici eludendo questioni più profonde sul potere e sulla responsabilità. Mettiamo in guardia contro questa tentazione." Piuttosto che fornire una panoramica dell’approccio FAccT all’etica (che è meglio fare in libri come quello), la nostra attenzione qui sarà sui limiti di questo tipo di inquadramento ristretto.

Un ottimo modo per valutare se una lente etica è completa è provare a fornire un esempio in cui la lente e le nostre intuizioni etiche danno risultati divergenti. Os Keyes, Jevan Hutson e Meredith Durbin hanno esplorato questo aspetto in modo grafico nel loro articolo ["A Mulching Proposal:
Analisi e miglioramento di un sistema algoritmico per trasformare gli anziani in liquami ad alto contenuto di nutrienti"](https://arxiv.org/abs/1908.06166). L'abstract del documento dice:

>: Le implicazioni etiche dei sistemi algoritmici sono state ampiamente discusse sia nell'HCI che nella comunità più ampia di coloro che sono interessati alla progettazione, allo sviluppo e alle politiche tecnologiche. In questo articolo esploriamo l’applicazione di un importante quadro etico – Equità, Responsabilità e Trasparenza – a un algoritmo proposto che risolve varie questioni sociali relative alla sicurezza alimentare e all’invecchiamento della popolazione. Utilizzando varie forme standardizzate di audit e valutazione algoritmica, aumentiamo drasticamente l'aderenza dell'algoritmo al quadro FAT, risultando in un sistema più etico e benefico. Discutiamo di come questo potrebbe servire da guida per altri ricercatori o professionisti che cercano di garantire migliori risultati etici dai sistemi algoritmici nella loro linea di lavoro.

In questo articolo, la proposta piuttosto controversa ("Trasformare gli anziani in liquami ad alto contenuto di nutrienti") e i risultati ("aumentare drasticamente l'aderenza dell'algoritmo al quadro FAT, risultando in un sistema più etico e benefico") sono in disaccordo... per non dire altro!

In filosofia, e in particolare in filosofia dell'etica, questo è uno degli strumenti più efficaci: in primo luogo, elaborare un processo, una definizione, una serie di domande, ecc., Progettato per risolvere qualche problema. Quindi prova a trovare un esempio in cui quella soluzione apparente si traduce in una proposta che nessuno considererebbe accettabile. Ciò può quindi portare ad un ulteriore perfezionamento della soluzione.

Finora ci siamo concentrati sulle cose che tu e la tua organizzazione potete fare. Ma a volte l’azione individuale o organizzativa non è sufficiente. Talvolta i governi devono anche considerare le implicazioni politiche.

## Ruolo della politica

Parliamo spesso con persone desiderose di soluzioni tecniche o di progettazione che rappresentino una soluzione completa al tipo di problemi di cui abbiamo discusso; ad esempio, un approccio tecnico ai dati sui debias o linee guida di progettazione per rendere la tecnologia meno avvincente. Sebbene tali misure possano essere utili, non saranno sufficienti per affrontare i problemi di fondo che hanno portato alla situazione attuale. Ad esempio, finché sarà incredibilmente redditizio creare tecnologie che creano dipendenza, le aziende continueranno a farlo, indipendentemente dal fatto che ciò abbia l’effetto collaterale di promuovere teorie del complotto e di inquinare il nostro ecosistema informativo. Sebbene i singoli designer possano provare a modificare la progettazione dei prodotti, non vedremo cambiamenti sostanziali finché non cambieranno gli incentivi al profitto sottostanti.

### L'efficacia della regolamentazione

Per vedere cosa può spingere le aziende ad agire concretamente, consideriamo i seguenti due esempi di come si è comportato Facebook. Nel 2018, un'indagine delle Nazioni Unite ha scoperto che Facebook aveva svolto un "ruolo determinante" nel genocidio in corso dei Rohingya, una minoranza etnica del Myanmar descritta dal segretario generale delle Nazioni Unite Antonio Guterres come "una delle persone, se non la, più discriminate al mondo". Gli attivisti locali avevano avvertito i dirigenti di Facebook che la loro piattaforma veniva utilizzata per diffondere discorsi di odio e incitare alla violenza già nel 2013. Nel 2015, erano stati avvertiti che Facebook avrebbe potuto svolgere in Myanmar lo stesso ruolo svolto dalle trasmissioni radiofoniche durante il genocidio ruandese (dove furono uccise un milione di persone). Eppure, alla fine del 2015, Facebook impiegava solo quattro collaboratori che parlavano birmano. Come ha detto una persona vicina alla questione: "Non è 20/20 col senno di poi. La portata di questo problema era significativa ed era già evidente". Zuckerberg ha promesso durante le udienze del Congresso di assumere “dozzine” per affrontare il genocidio in Myanmar (nel 2018, anni dopo l’inizio del genocidio, compresa la distruzione mediante incendio di almeno 288 villaggi nello stato settentrionale di Rakhine dopo l’agosto 2017).

Ciò è in netto contrasto con la rapida assunzione di 1.200 persone in Germania da parte di Facebook(http://thehill.com/policy/technology/361722-facebook-opens-second-german-office-to-comply-with-hate-speech-law) nel tentativo di evitare sanzioni costose (fino a 50 milioni di euro) ai sensi della nuova legge tedesca contro l'incitamento all'odio. Chiaramente, in questo caso, Facebook è stato più reattivo alla minaccia di una sanzione finanziaria che alla distruzione sistematica di una minoranza etnica.

In un [articolo sulle questioni relative alla privacy](https://idlewords.com/2019/06/the_new_wilderness.htm), Maciej Ceglowski traccia paralleli con il movimento ambientalista: 

> : Questo progetto normativo ha avuto così tanto successo nel Primo Mondo che rischiamo di dimenticare com'era la vita prima. Lo smog soffocante del tipo che oggi uccide migliaia di persone a Giakarta e Delhi era https://en.wikipedia.org/wiki/Pea_soup_fog[una volta emblematico di Londra]. Il fiume Cuyahoga in Ohio era solito http://www.ohiohistorycentral.org/w/Cuyahoga_River_Fire[prendere fuoco in modo affidabile]. In un esempio particolarmente orribile di conseguenze impreviste, il piombo tetraetile aggiunto alla benzina https://en.wikipedia.org/wiki/Lead%E2%80%93crime_hypothesis[ha aumentato i tassi di criminalità violenta] in tutto il mondo per cinquant'anni. Nessuno di questi danni avrebbe potuto essere risolto dicendo alle persone di votare con il portafoglio, o rivedendo attentamente le politiche ambientali di ogni azienda a cui avevano affidato la propria attività, o di smettere di usare le tecnologie in questione. Per risolverli è stata necessaria una regolamentazione coordinata, e talvolta altamente tecnica, oltre i confini giurisdizionali. In alcuni casi, come il https://en.wikipedia.org/wiki/Montreal_Protocol[divieto dei refrigeranti commerciali] che ha ridotto lo strato di ozono, tale regolamento richiedeva un consenso mondiale. Siamo al punto in cui abbiamo bisogno di un simile cambiamento di prospettiva nella nostra legge sulla privacy.

### Diritti e politica

L’aria pulita e l’acqua potabile pulita sono beni pubblici che è quasi impossibile proteggere attraverso decisioni di mercato individuali, ma richiedono piuttosto un’azione normativa coordinata. Allo stesso modo, molti dei danni derivanti da conseguenze involontarie di abusi della tecnologia riguardano beni pubblici, come un ambiente informativo inquinato o un deterioramento della privacy ambientale. Troppo spesso la privacy è considerata un diritto individuale, ma la sorveglianza diffusa ha conseguenze sociali (cosa che varrebbe anche se fosse possibile per alcuni individui rinunciarvi).

Molti dei problemi che stiamo riscontrando nella tecnologia sono in realtà problemi di diritti umani, come quando un algoritmo parziale raccomanda che gli imputati neri abbiano pene detentive più lunghe, quando particolari annunci di lavoro vengono mostrati solo ai giovani o quando la polizia usa il riconoscimento facciale per identificare i manifestanti. La sede appropriata per affrontare le questioni legate ai diritti umani è generalmente la legge.

Abbiamo bisogno di cambiamenti sia normativi che legali, *e* del comportamento etico degli individui. Il cambiamento del comportamento individuale non può affrontare incentivi di profitto disallineati, esternalità (dove le aziende raccolgono grandi profitti scaricando i costi e i danni sulla società nel suo complesso) o fallimenti sistemici. Tuttavia, la legge non coprirà mai tutti i casi limite ed è importante che i singoli sviluppatori di software e data scientist siano attrezzati per prendere decisioni etiche nella pratica.

### Automobili: un precedente storico

I problemi che dobbiamo affrontare sono complessi e non esistono soluzioni semplici. Ciò può essere scoraggiante, ma troviamo speranza nel considerare altre grandi sfide che le persone hanno affrontato nel corso della storia. Un esempio è il movimento per aumentare la sicurezza automobilistica, trattato come caso di studio in ["Datasheets for Datasets"](https://arxiv.org/abs/1803.09010) di Timnit Gebru et al. e nel podcast sul design [99% Invisible](https://99percentinvisible.org/episode/nut-behind-wheel/). Le prime auto non avevano cinture di sicurezza, manopole di metallo sul cruscotto che potevano conficcarsi nel cranio delle persone durante un incidente, normali finestrini in vetro che si frantumavano in modi pericolosi e piantoni dello sterzo non pieghevoli che impalavano i conducenti. Tuttavia, le case automobilistiche erano incredibilmente restie anche solo a discutere l’idea di sicurezza come qualcosa che potevano contribuire ad affrontare, e la convinzione diffusa era che le auto fossero proprio come sono e che fossero le persone che le usavano a causare problemi.

Ci sono voluti decenni di lavoro da parte degli attivisti e dei sostenitori della sicurezza dei consumatori per cambiare il dibattito nazionale e considerare che forse le case automobilistiche avevano delle responsabilità che dovrebbero essere affrontate attraverso la regolamentazione. Quando fu inventato il piantone dello sterzo pieghevole, non fu implementato per diversi anni poiché non vi erano incentivi finanziari per farlo. La principale casa automobilistica General Motors ha assunto investigatori privati ​​per cercare di portare alla luce il caso del difensore della sicurezza dei consumatori Ralph Nader. L'esigenza di cinture di sicurezza, manichini per crash test e piantoni dello sterzo pieghevoli sono state vittorie importanti. È stato solo nel 2011 che le case automobilistiche hanno dovuto iniziare a utilizzare manichini per crash test che rappresentassero il corpo della donna media, e non solo il corpo dell’uomo medio; prima di questo, le donne avevano il 40% in più di probabilità di rimanere ferite in un incidente stradale di pari impatto rispetto a un uomo. Questo è un vivido esempio dei modi in cui pregiudizi, politiche e tecnologia hanno conseguenze importanti.

## Conclusione

Provenendo da un background di lavoro con la logica binaria, la mancanza di risposte chiare in etica può essere inizialmente frustrante.  Tuttavia, le implicazioni dell’impatto del nostro lavoro sul mondo, comprese le conseguenze indesiderate e il fatto che il lavoro venga utilizzato come arma da cattivi attori, sono alcune delle domande più importanti che possiamo (e dovremmo!) considerare.  Anche se non ci sono risposte facili, ci sono alcune trappole da evitare e pratiche da seguire per procedere verso un comportamento più etico.

Molte persone (compresi noi!) sono alla ricerca di risposte più soddisfacenti e concrete su come affrontare gli impatti dannosi della tecnologia. Tuttavia, data la natura complessa, di vasta portata e interdisciplinare dei problemi che stiamo affrontando, non esistono soluzioni semplici. Julia Angwin, ex reporter senior di ProPublica che si concentra su questioni di bias algoritmico e sorveglianza (e uno dei ricercatori del 2016 dell'algoritmo di recidività COMPAS che ha contribuito a innescare il campo di FAccT) ha detto in [a 2019 intervista](https://www.fastcompany.com/90337954/who-cares-about-liberty-julia-angwin-and-trevor-paglen-on-privacy-surveillance-and-the-mess-were-in):

> : Credo fortemente che per risolvere un problema sia necessario diagnosticarlo, e che su questo siamo ancora nella fase di diagnosi. Se si pensa alla fine del secolo e all’industrializzazione, abbiamo avuto, non so, 30 anni di lavoro minorile, orari di lavoro illimitati, condizioni di lavoro terribili, e ci sono volute un sacco di critiche e di sostegno da parte dei giornalisti per diagnosticare il problema e capire di cosa si trattasse, e poi l’attivismo per cambiare le leggi. Mi sento come se fossimo in una seconda industrializzazione delle informazioni sui dati... Vedo il mio ruolo nel cercare di rendere il più chiaro possibile quali sono gli aspetti negativi e nel diagnosticarli in modo molto accurato in modo che possano essere risolvibili. È un lavoro duro e molte più persone devono farlo. 

È rassicurante che Angwin pensi che siamo ancora in gran parte nella fase di diagnosi: se la comprensione di questi problemi sembra incompleta, è normale e naturale. Nessuno ha ancora una “cura”, anche se è fondamentale continuare a lavorare per comprendere e affrontare meglio i problemi che stiamo affrontando.

Uno dei nostri revisori di questo libro, Fred Monroe, lavorava nel trading di hedge fund. Ci ha detto, dopo aver letto questo capitolo, che molte delle questioni discusse qui (la distribuzione dei dati è radicalmente diversa da quella su cui è stato addestrato un modello, i cicli di feedback dell’impatto su un modello una volta implementato e su larga scala, e così via) erano anche questioni chiave per la costruzione di modelli di trading redditizi. Il tipo di cose che devi fare per considerare le conseguenze sociali si sovrapporranno molto con le cose che devi fare per considerare le conseguenze a livello organizzativo, di mercato e sui clienti, quindi pensare attentamente all’etica può anche aiutarti a pensare attentamente a come rendere il tuo prodotto dati di successo più in generale!
