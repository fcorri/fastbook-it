## Reti neurali: una breve storia

Nel 1943 Warren McCulloch, un neurofisiologo, e Walter Pitts, un logico, collaborarono per sviluppare un modello matematico di un neurone artificiale. Nel loro [articolo](https://link.springer.com/article/10.1007/BF02478259) "Un calcolo logico delle idee immanenti nell'attività nervosa" hanno dichiarato che:

> : A causa del carattere “tutto o niente” dell'attività nervosa, gli eventi neurali e le relazioni tra loro possono essere trattati mediante la logica proposizionale. Si è constatato che il comportamento di ogni rete può essere descritto in questi termini.

McCulloch e Pitts si resero conto che un modello semplificato di un neurone reale poteva essere rappresentato utilizzando semplici addizioni e soglie, come mostrato in <<neurone>>. Pitts era autodidatta e all'età di 12 anni aveva ricevuto un'offerta per studiare all'Università di Cambridge con il grande Bertrand Russell. Non accettò questo invito, e anzi per tutta la sua vita non accettò offerte di gradi avanzati o di incarichi di autorità. La maggior parte del suo famoso lavoro è stato realizzato mentre era senza casa. Nonostante la mancanza di una posizione ufficialmente riconosciuta e il crescente isolamento sociale, il suo lavoro con McCulloch fu influente e fu intrapreso da uno psicologo di nome Frank Rosenblatt.

<img alt="Neuroni naturali e artificiali" width="500" caption="Neuroni naturali e artificiali" src="images/chapter7_neuron.png" id="neurone"/>

Rosenblatt sviluppò ulteriormente il neurone artificiale per conferirgli la capacità di apprendere. Ancora più importante, lavorò alla costruzione del primo dispositivo che utilizzava effettivamente questi principi, il Mark I Perceptron. In "La progettazione di un automatismo intelligente" Rosenblatt ha scritto di questo lavoro: "Stiamo per assistere alla nascita di una macchina del genere, una macchina capace di percepire, riconoscere e identificare l'ambiente circostante senza alcun addestramento o controllo umano". Il perceptron è stato costruito ed è stato in grado di riconoscere con successo forme semplici.

Un professore del MIT di nome Marvin Minsky (che era un anno dietro Rosenblatt nella stessa scuola superiore!), insieme a Seymour Papert, scrisse un libro intitolato _Perceptrons_ (MIT Press), sull'invenzione di Rosenblatt. Hanno dimostrato che un singolo strato di questi dispositivi non era in grado di apprendere alcune funzioni matematiche semplici ma critiche (come XOR). Nello stesso libro hanno anche dimostrato che l’utilizzo di più livelli di dispositivi consentirebbe di risolvere queste limitazioni. Sfortunatamente, solo la prima di queste intuizioni è stata ampiamente riconosciuta. Di conseguenza, la comunità accademica globale rinunciò quasi completamente alle reti neurali per i successivi due decenni.

Forse il lavoro più importante nel campo delle reti neurali negli ultimi 50 anni è stato il *Parallel Distributed Processing* (PDP) in più volumi di David Rumelhart, James McClellan e il PDP Research Group, pubblicato nel 1986 da MIT Press. Il capitolo 1 espone una speranza simile a quella mostrata da Rosenblatt:

> : Le persone sono più intelligenti dei computer di oggi perché il cervello utilizza un'architettura computazionale di base più adatta a gestire un aspetto centrale dei compiti naturali di elaborazione delle informazioni in cui le persone sono così brave. ...Introdurremo un quadro computazionale per modellare i processi cognitivi che sembra... più vicino di altri quadri allo stile di calcolo come potrebbe essere fatto dal cervello.

La premessa che PDP utilizza qui è che i programmi per computer tradizionali funzionano in modo molto diverso dal cervello, e questo potrebbe essere il motivo per cui i programmi per computer erano stati (a quel punto) così pessimi nel fare cose che il cervello trova facili (come riconoscere gli oggetti nelle immagini). Gli autori hanno affermato che l'approccio PDP era "più vicino 
rispetto ad altre strutture" su come funziona il cervello, e quindi potrebbe essere in grado di gestire meglio questo tipo di compiti.

In effetti, l'approccio presentato in PDP è molto simile all'approccio utilizzato nelle odierne reti neurali. Il libro definisce l'elaborazione distribuita parallela come richiedente:

1. Un insieme di *unità di elaborazione*
1. Uno *stato di attivazione*
1. Una *funzione di uscita* per ciascuna unità 
1. Un *modello di connettività* tra le unità 
1. Una *regola di propagazione* per propagare modelli di attività attraverso la rete di connettività 
1. Una *regola di attivazione* per combinare gli input che influiscono su un'unità con lo stato attuale di quell'unità per produrre un output per l'unità
1. Una *regola di apprendimento* in base alla quale i modelli di connettività vengono modificati dall'esperienza 
1. Un *ambiente* all'interno del quale il sistema deve operare

Vedremo in questo libro che le moderne reti neurali gestiscono ciascuno di questi requisiti.

Negli anni '80 la maggior parte dei modelli erano costruiti con un secondo strato di neuroni, evitando così il problema identificato da Minsky e Papert (questo era il loro "modello di connettività tra le unità", per usare il quadro sopra). E in effetti, le reti neurali sono state ampiamente utilizzate negli anni '80 e '90 per progetti reali e pratici. Tuttavia, ancora una volta un malinteso sulle questioni teoriche ha frenato il campo. In teoria, aggiungere solo uno strato extra di neuroni era sufficiente per consentire l’approssimazione di qualsiasi funzione matematica con queste reti neurali, ma in pratica tali reti erano spesso troppo grandi e troppo lente per essere utili.

Sebbene i ricercatori abbiano dimostrato 30 anni fa che per ottenere buone prestazioni pratiche è necessario utilizzare ancora più strati di neuroni, è solo nell’ultimo decennio che questo principio è stato maggiormente apprezzato e applicato. Le reti neurali sono ora finalmente all’altezza del loro potenziale, grazie all’uso di più livelli, unito alla capacità di farlo grazie ai miglioramenti nell’hardware del computer, all’aumento della disponibilità dei dati e alle modifiche algoritmiche che consentono alle reti neurali di essere addestrate più velocemente e più facilmente. Ora abbiamo ciò che Rosenblatt aveva promesso: "una macchina capace di percepire, riconoscere e identificare l'ambiente circostante senza alcun addestramento o controllo umano".

Questo è ciò che imparerai a costruire in questo libro. Ma prima, visto che passeremo molto tempo insieme, conosciamoci un po'...

## Come apprendere il deep learning

Il professore di Harvard David Perkins, autore di _Making Learning Whole_ (Jossey-Bass), ha molto da dire sull'insegnamento. L'idea di base è insegnare *l'intero gioco*. Ciò significa che se insegni baseball, prima porti le persone a una partita di baseball o le fai giocare. Non insegni loro come avvolgere lo spago per creare una palla da baseball da zero, la fisica di una parabola o il coefficiente di attrito di una palla su una mazza.

Paul Lockhart, un dottorato in matematica della Columbia, ex professore alla Brown e insegnante di matematica K-12, immagina nell'influente [saggio](https://www.maa.org/external_archive/devlin/LockhartsLament.pdf) "A Mathematician's Lament" un mondo da incubo in cui la musica e l'arte vengono insegnate nello stesso modo in cui viene insegnata la matematica. Ai bambini non è consentito ascoltare o suonare musica finché non hanno trascorso più di un decennio a padroneggiare la notazione e la teoria musicale, frequentando lezioni di trasposizione di spartiti in una chiave diversa. Nelle lezioni d'arte, gli studenti studiano colori e applicatori, ma non possono dipingere fino al college. Sembra assurdo? Questo è il modo in cui viene insegnata la matematica: richiediamo agli studenti di dedicare anni alla memorizzazione meccanica e all'apprendimento di *fondamenti* aridi e sconnessi che affermiamo verranno ripagati in seguito, molto tempo dopo che la maggior parte di loro avrà abbandonato la materia.

Sfortunatamente, è qui che iniziano molte risorse didattiche sul deep learning, chiedendo agli studenti di seguire la definizione dell'Assia e i teoremi per l'approssimazione di Taylor delle funzioni di perdita, senza mai fornire esempi di codice effettivamente funzionante. Non stiamo battendo i calcoli. Adoriamo il calcolo infinitesimale e Sylvain lo ha insegnato anche a livello universitario, ma non pensiamo che sia il posto migliore da cui iniziare quando si impara il deep learning!

Nel deep learning, è davvero utile avere la motivazione per correggere il modello per farlo funzionare meglio. È allora che inizi ad apprendere la teoria pertinente. Ma prima devi avere il modello. Insegniamo quasi tutto attraverso esempi reali. Mentre costruiamo questi esempi, andiamo sempre più in profondità e ti mostreremo come rendere i tuoi progetti sempre migliori. Ciò significa che apprenderai gradualmente tutte le basi teoriche di cui hai bisogno, contestualizzate, in modo tale da capire perché è importante e come funziona.

Quindi, ecco il nostro impegno nei tuoi confronti. In questo libro seguiremo questi principi:

- Insegnare l'*intero gioco*. Inizieremo mostrando come utilizzare una rete di deep learning completa, funzionante, molto utilizzabile e all'avanguardia per risolvere problemi del mondo reale, utilizzando strumenti semplici ed espressivi. E poi gradualmente scaveremo sempre più in profondità per capire come sono realizzati questi strumenti, e come sono realizzati gli strumenti che li producono, e così via...
- Insegnare sempre attraverso esempi. Faremo in modo che ci sia un contesto e uno scopo che tu possa comprendere in modo intuitivo, piuttosto che iniziare con la manipolazione dei simboli algebrici.
- Semplificare il più possibile. Abbiamo trascorso anni a costruire strumenti e metodi di insegnamento che rendono molto semplici argomenti precedentemente complessi.
- Rimozione delle barriere. Finora il deep learning è stato un gioco molto esclusivo. Lo stiamo aprendo e assicurando che tutti possano giocare.

Abbiamo completato centinaia di progetti di machine learning utilizzando decine di pacchetti diversi e molti linguaggi di programmazione diversi. I primi due passaggi devono essere eseguiti solo una volta sul tuo server GPU. Se esegui nuovamente la cella, utilizzerà il set di dati e il modello che sono già stati scaricati, anziché scaricarli nuovamente. Diamo un'occhiata al contenuto della cella e ai risultati (<<first_training>>):

````python
#id first_training
#caption Results from the first training
# CLICK ME
from fastai.vision.all import *
path = untar_data(URLs.PETS)/'images'

def is_cat(x): return x[0].isupper()
dls = ImageDataLoaders.from_name_func(
    path, get_image_files(path), valid_pct=0.2, seed=42,
    label_func=is_cat, item_tfms=Resize(224))

learn = vision_learner(dls, resnet34, metrics=error_rate)
learn.fine_tune(1)
````

Probabilmente non vedrai esattamente gli stessi risultati che sono nel libro. Ci sono molte fonti di piccole variazioni casuali coinvolte nei modelli di addestramento. Tuttavia in questo esempio generalmente vediamo un tasso di errore ben inferiore a 0,02.

### Che cos'è l'apprendimento automatico?

Il tuo classificatore è un modello di deep learning. Come già accennato, i modelli di deep learning utilizzano reti neurali, che risalgono originariamente agli anni ’50 e sono diventate potenti solo di recente grazie ai recenti progressi.

Un altro elemento chiave del contesto è che il deep learning è solo un’area moderna nella disciplina più generale del *machine learning*. Per comprendere l'essenza di ciò che hai fatto quando hai addestrato il tuo modello di classificazione, non è necessario comprendere il deep learning. È sufficiente vedere come il tuo modello e il tuo processo di formazione siano esempi dei concetti che si applicano all'apprendimento automatico in generale.

Quindi in questa sezione descriveremo cos'è l'apprendimento automatico. Esamineremo i concetti chiave e mostreremo come possono essere ricondotti al saggio originale che li ha introdotti.

Il *machine learning* è, come la normale programmazione, un modo per far sì che i computer completino un'attività specifica. Ma come utilizzeremmo la programmazione regolare per fare ciò che abbiamo appena fatto nell'ultima sezione: riconoscere i cani dai gatti nelle foto? Ma riconoscere gli oggetti in una foto è un po' complicato; quali *sono* i passi che compiamo quando riconosciamo un oggetto in una foto? Non lo sappiamo davvero, poiché tutto avviene nel nostro cervello senza che ne siamo consapevoli!

Già agli albori dell’informatica, nel 1949, un ricercatore IBM di nome Arthur Samuel iniziò a lavorare su un modo diverso di far eseguire ai computer i compiti, che chiamò *apprendimento automatico*. Nel suo saggio classico del 1962 "Intelligenza Artificiale: una frontiera dell'automazione", scrisse:

>: Programmare un computer per tali calcoli è, nella migliore delle ipotesi, un compito difficile, non principalmente a causa della complessità intrinseca del computer stesso ma, piuttosto, a causa della necessità di descrivere ogni minimo passaggio del processo nei dettagli più esasperanti. I computer, come ti dirà qualsiasi programmatore, sono degli idioti giganti, non dei cervelli giganti.

La sua idea di base era questa: invece di dire al computer i passaggi esatti necessari per risolvere un problema, mostrargli esempi del problema da risolvere e lasciare che sia lui a capire come risolverlo da solo. Ciò si rivelò molto efficace: nel 1961 il suo programma di gioco della dama aveva imparato così tanto da battere il campione dello stato del Connecticut! Ecco come ha descritto la sua idea (dallo stesso saggio di cui sopra):

>: Supponiamo di predisporre alcuni mezzi automatici per testare l'efficacia di qualsiasi attuale assegnazione di peso in termini di prestazione effettiva e di fornire un meccanismo per alterare l'assegnazione di peso in modo da massimizzare la prestazione. Non abbiamo bisogno di entrare nei dettagli di tale procedura per vedere che potrebbe essere resa completamente automatica e per vedere che una macchina così programmata "impararebbe" dalla sua esperienza.

Ci sono una serie di concetti potenti racchiusi in questa breve affermazione: 

- L'idea di una "assegnazione del peso" 
- Il fatto che ogni assegnazione di peso abbia una "prestazione effettiva"
- L'esigenza che esista un "mezzo automatico" per testare tale prestazione,  
- La necessità di un "meccanismo" (cioè un altro processo automatico) per migliorare le prestazioni modificando le assegnazioni di peso

Prendiamo in esame questi concetti uno per uno, per capire come si incastrano nella pratica. Per prima cosa dobbiamo capire cosa intende Samuel con *assegnazione di peso*.

I pesi sono solo variabili e un'assegnazione di peso è una scelta particolare di valori per tali variabili. Gli input del programma sono valori che elabora per produrre i suoi risultati, ad esempio prendendo i pixel dell'immagine come input e restituendo come risultato la classificazione "cane". Le assegnazioni di peso del programma sono altri valori che definiscono come funzionerà il programma.

Poiché influenzeranno il programma, in un certo senso sono un altro tipo di input, quindi aggiorneremo la nostra immagine di base in <<basic_program>> e la sostituiremo con <<weight_assignment>> per tenerne conto.

````python
#hide_input
#caption A program using weight assignment
#id weight_assignment
gv('''model[shape=box3d width=1 height=0.7]
inputs->model->results; weights->model''')
````

Abbiamo cambiato il nome del nostro box da *programma* a *modello*. Questo per seguire la terminologia moderna e per riflettere che il *modello* è un tipo speciale di programma: è uno che può fare *molte cose diverse*, a seconda dei *pesi*. Può essere implementato in molti modi diversi. Ad esempio, nel programma di dama di Samuel, valori diversi dei pesi darebbero come risultato diverse strategie di gioco della dama. 

(A proposito, quelli che Samuel chiamava "pesi" oggigiorno vengono generalmente definiti *parametri* del modello, nel caso in cui tu abbia incontrato quel termine. Il termine *pesi* è riservato a un particolare tipo di parametro del modello.)

Successivamente, Samuel ha affermato che abbiamo bisogno di un *mezzo automatico per testare l'efficacia di qualsiasi assegnazione di peso attuale in termini di prestazioni effettive*. Nel caso del suo programma di dama, la "prestazione effettiva" di un modello sarebbe quanto bene gioca. E potresti testare automaticamente le prestazioni di due modelli impostandoli per giocare uno contro l'altro e vedendo quale di solito vince.

Infine, afferma che abbiamo bisogno di *un meccanismo per alterare l'assegnazione del peso in modo da massimizzare la prestazione*. Ad esempio, potremmo considerare la differenza di peso tra il modello vincente e quello perdente, e aggiustare ulteriormente i pesi nella direzione vincente.

Ora possiamo capire perché disse che tale procedura *potrebbe essere resa completamente automatica e... una macchina così programmata "impararebbe" dalla sua esperienza*. L’apprendimento diventerebbe completamente automatico quando anche la regolazione dei pesi fosse automatica: quando invece di migliorare un modello regolandone manualmente i pesi, ci affidavamo a un meccanismo automatizzato che produceva aggiustamenti basati sulle prestazioni.

<<training_loop>> mostra il quadro completo dell'idea di Samuel di addestrare un modello di machine learning.

````python
#hide_input
#caption Training a machine learning model
#id training_loop
#alt The basic training loop
gv('''ordering=in
model[shape=box3d width=1 height=0.7]
inputs->model->results; weights->model; results->performance
performance->weights[constraint=false label=update]''')
````

Notare la distinzione tra i *risultati* del modello (ad esempio, le mosse in una partita a dama) e le sue *prestazioni* (ad esempio, se vince la partita o quanto velocemente vince). 

Tieni inoltre presente che una volta addestrato il modello, ovvero una volta che abbiamo scelto la nostra assegnazione di peso finale, migliore e preferita, allora possiamo pensare ai pesi come *parte del modello*, poiché non li stiamo più variando.

### Cos'è una rete neurale?

Non è troppo difficile immaginare come potrebbe apparire il modello per un programma di dama. Potrebbero esserci una serie di strategie di pedina codificate e qualche tipo di meccanismo di ricerca, quindi i pesi potrebbero variare come vengono selezionate le strategie, su quali parti del tabellone si concentrano durante una ricerca e così via. Ma non è affatto ovvio come potrebbe apparire il modello per un programma di riconoscimento delle immagini, o per la comprensione del testo, o per molti altri problemi interessanti che potremmo immaginare.

Ciò che vorremmo è una sorta di funzione così flessibile da poter essere utilizzata per risolvere qualsiasi problema, semplicemente variandone il peso. Sorprendentemente, questa funzione esiste davvero! È la rete neurale, di cui abbiamo già parlato. Cioè, se consideriamo una rete neurale come una funzione matematica, risulta essere una funzione estremamente flessibile a seconda dei suoi pesi. Una dimostrazione matematica chiamata *teorema di approssimazione universale* mostra che questa funzione può risolvere qualsiasi problema a qualsiasi livello di precisione, in teoria. Il fatto che le reti neurali siano così flessibili significa che, in pratica, sono spesso un tipo di modello adatto e puoi concentrare i tuoi sforzi sul processo di addestramento, ovvero sull’individuazione di buone assegnazioni di peso.

Ma che dire di questo processo?  Si potrebbe immaginare che potrebbe essere necessario trovare un nuovo "meccanismo" per aggiornare automaticamente i pesi per ogni problema. Sarebbe laborioso. Ciò che vorremmo anche qui è un modo del tutto generale per aggiornare i pesi di una rete neurale, per farla migliorare in ogni determinato compito. Convenientemente esiste anche questo!

Questo si chiama *discesa del gradiente stocastico* (SGD). Vedremo in dettaglio come funzionano le reti neurali e l'SGD in <<chapter_mnist_basics>>, oltre a spiegare il teorema di approssimazione universale. Per ora, tuttavia, utilizzeremo invece le parole di Samuel: *Non abbiamo bisogno di entrare nei dettagli di tale procedura per vedere che potrebbe essere resa completamente automatica e per vedere che una macchina così programmata "impara" dalla sua esperienza.*

> J: Non preoccuparti, né SGD né le reti neurali sono matematicamente complesse. Entrambi si affidano quasi interamente all'addizione e alla moltiplicazione per svolgere il loro lavoro (ma fanno _molte_ addizioni e moltiplicazioni!). La reazione principale che sentiamo dagli studenti quando vedono i dettagli è: "È tutto?"

### Limitazioni inerenti al machine learning

Da questa immagine ora possiamo vedere alcune cose fondamentali sull'addestramento di un modello di deep learning:

- Non è possibile creare un modello senza dati.
- Un modello può imparare a operare solo sui modelli visti nei dati di input utilizzati per addestrarlo.
- Questo approccio di apprendimento crea solo *previsioni*, non *azioni* consigliate.
- Non è sufficiente avere solo esempi di dati di input; abbiamo bisogno di *etichette* anche per quei dati (ad esempio, le immagini di cani e gatti non sono sufficienti per addestrare un modello; abbiamo bisogno di un'etichetta per ciascuno, indicando quali sono cani e quali sono gatti).

In generale, abbiamo visto che la maggior parte delle organizzazioni che affermano di non disporre di dati sufficienti, in realtà intendono dire di non disporre di sufficienti dati *etichettati*. Se un'organizzazione è interessata a fare qualcosa nella pratica con un modello, allora presumibilmente ha degli input rispetto ai quali intende eseguire il proprio modello. E presumibilmente lo fanno in altro modo da un po' (ad esempio, manualmente o con qualche programma euristico), quindi hanno dati da quei processi! Ad esempio, uno studio di radiologia avrà quasi certamente un archivio di scansioni mediche (poiché devono essere in grado di verificare come i loro pazienti stanno progredendo nel tempo), ma tali scansioni potrebbero non avere etichette strutturate contenenti un elenco di diagnosi o interventi (poiché i radiologi generalmente creano rapporti in linguaggio naturale a testo libero, non dati strutturati). Discuteremo molto degli approcci all'etichettatura in questo libro, perché è una questione molto importante nella pratica.

Poiché questi tipi di modelli di machine learning possono solo fare *previsioni* (ovvero tentare di replicare le etichette), ciò può comportare un divario significativo tra gli obiettivi organizzativi e le capacità del modello. Ad esempio, in questo libro imparerai come creare un *sistema di consigli* in grado di prevedere quali prodotti un utente potrebbe acquistare. Viene spesso utilizzato nell'e-commerce, ad esempio per personalizzare i prodotti mostrati in una home page mostrando gli articoli con il punteggio più alto. Ma un modello di questo tipo viene generalmente creato osservando un utente e la sua cronologia degli acquisti (*input*) e cosa ha acquistato o guardato (*etichette*), il che significa che è probabile che il modello ti parli di prodotti che l'utente ha già o di cui è già a conoscenza, piuttosto che nuovi prodotti di cui è molto probabile che sia interessato a sentire. È molto diverso da quello che potrebbe fare, ad esempio, un esperto del tuo libraio locale, dove fa domande per capire i tuoi gusti e poi ti parla di autori o serie di cui non hai mai sentito parlare prima.

Un’altra intuizione fondamentale deriva dalla considerazione di come un modello interagisce con il suo ambiente. Questo può creare *cicli di feedback*, come descritto qui:

- Viene creato un modello di *polizia predittiva* basato sui luoghi in cui sono stati effettuati gli arresti in passato. In pratica, ciò non prevede realmente la criminalità, ma piuttosto gli arresti, e quindi in parte riflette semplicemente i pregiudizi nei processi di polizia esistenti.
- Gli agenti delle forze dell'ordine potrebbero quindi utilizzare quel modello per decidere dove concentrare la propria attività di polizia, con conseguente aumento degli arresti in quelle aree.
- I dati su questi ulteriori arresti verrebbero poi reinseriti per riqualificare le future versioni del modello.

Si tratta di un *ciclo di feedback positivo*, in cui più viene utilizzato il modello, più i dati diventano distorti, rendendo il modello ancora più distorto e così via.

I cicli di feedback possono anche creare problemi in contesti commerciali. Ad esempio, un sistema di raccomandazione video potrebbe essere sbilanciato nel consigliare i contenuti consumati dai più grandi spettatori di video (ad esempio, i teorici della cospirazione e gli estremisti tendono a guardare più contenuti video online rispetto alla media), con il risultato che quegli utenti aumentano il loro consumo di video, con il risultato che vengono consigliati più di questi tipi di video. Tratteremo questo argomento più in dettaglio nel <<capitolo_etica>>.

Esistono varie classi diverse per diversi tipi di set di dati e problemi di deep learning: qui utilizziamo "ImageDataLoaders". La prima parte del nome della classe sarà generalmente il tipo di dati che hai, come immagine o testo.

L'altra informazione importante che dobbiamo comunicare a Fastai è come ottenere le etichette dal set di dati. I set di dati di visione artificiale sono normalmente strutturati in modo tale che l'etichetta di un'immagine sia parte del nome del file o del percorso, più comunemente il nome della cartella principale. fastai viene fornito con una serie di metodi di etichettatura standardizzati e modi per scriverne di propri. Qui stiamo dicendo a fastai di usare la funzione `is_cat` che abbiamo appena definito.

Infine, definiamo le "trasformazioni" di cui abbiamo bisogno. Una "trasformazione" contiene codice che viene applicato automaticamente durante l'addestramento; fastai include molti `Transform` predefiniti e aggiungerne di nuovi è semplice come creare una funzione Python. Ce ne sono due tipi: `item_tfms` vengono applicati a ciascun elemento (in questo caso, ciascun elemento viene ridimensionato a un quadrato di 224 pixel), mentre `batch_tfms` vengono applicati a un *batch* di elementi alla volta utilizzando la GPU, quindi sono particolarmente veloci (ne vedremo molti esempi in questo libro).

Perché 224 pixel? Questa è la dimensione standard per ragioni storiche (i vecchi modelli preaddestrati richiedono esattamente questa dimensione), ma puoi passare praticamente qualsiasi cosa. Se aumenti le dimensioni, spesso otterrai un modello con risultati migliori (poiché sarà in grado di mettere a fuoco più dettagli), ma a scapito della velocità e del consumo di memoria; è vero il contrario se si diminuiscono le dimensioni.

> Nota: classificazione e regressione: _classificazione_ e _regressione_ hanno significati molto specifici nell'apprendimento automatico. Questi sono i due principali tipi di modello che esamineremo in questo libro. Un modello di classificazione è quello che tenta di prevedere una classe o categoria. Cioè, prevede una serie di possibilità discrete, come "cane" o "gatto". Un modello di regressione tenta di prevedere una o più quantità numeriche, come una temperatura o una posizione. A volte le persone usano la parola _regressione_ per riferirsi a un particolare tipo di modello chiamato _modello di regressione lineare_; questa è una cattiva pratica e non utilizzeremo quella terminologia in questo libro!

Il set di dati Pet contiene 7.390 immagini di cani e gatti, costituiti da 37 razze diverse. Ogni immagine è etichettata utilizzando il suo nome file: ad esempio il file *great\_pyrenees\_173.jpg* è il 173esimo esempio di immagine di un cane di razza Great Pyrenees nel set di dati. I nomi dei file iniziano con una lettera maiuscola se l'immagine è un gatto, altrimenti con una lettera minuscola. Dobbiamo dire a fastai come ottenere le etichette dai nomi dei file, cosa che facciamo chiamando `from_name_func` (il che significa che le etichette possono essere estratte utilizzando una funzione applicata al nome del file) e passando `is_cat`, che restituisce `x[0].isupper()`, che restituisce `True` se la prima lettera è maiuscola (cioè è un gatto).

Il parametro più importante da menzionare qui è "valid_pct=0.2". Questo dice a fastai di trattenere il 20% dei dati e di *non usarli affatto per addestrare il modello*. Questo 20% dei dati è chiamato *set di convalida*; il restante 80% è chiamato *set di allenamento*. Il set di validazione viene utilizzato per misurare l'accuratezza del modello. Per impostazione predefinita, il 20% trattenuto viene selezionato in modo casuale. Il parametro `seed=42` imposta il *seme casuale* sullo stesso valore ogni volta che eseguiamo questo codice, il che significa che otteniamo lo stesso set di convalida ogni volta che lo eseguiamo: in questo modo, se modifichiamo il nostro modello e lo riqualifichiamo, sappiamo che eventuali differenze sono dovute alle modifiche al modello, non all'avere un diverso set di convalida casuale.

fastai ti mostrerà *sempre* la precisione del tuo modello utilizzando *solo* il set di validazione, *mai* il set di addestramento. Questo è assolutamente fondamentale, perché se addestri un modello abbastanza grande per un tempo sufficientemente lungo, alla fine memorizzerà l'etichetta di ogni elemento nel tuo set di dati! Il risultato non sarà effettivamente un modello utile, perché ciò che ci interessa è quanto bene il nostro modello funzioni su *immagini mai viste*. Questo è sempre il nostro obiettivo quando creiamo un modello: che sia utile sui dati che il modello vede solo in futuro, dopo essere stato addestrato.

Anche quando il tuo modello non ha memorizzato completamente tutti i tuoi dati, nelle prime fasi dell'addestramento potrebbe averne memorizzato alcune parti. Di conseguenza, più a lungo ti alleni, migliore sarà la tua precisione nel set di allenamento; Anche l'accuratezza del set di validazione migliorerà per un po', ma alla fine inizierà a peggiorare quando il modello inizierà a memorizzare il set di addestramento, invece di trovare modelli sottostanti generalizzabili nei dati. Quando ciò accade, diciamo che il modello è *overfitting*.

Come puoi vedere, sebbene le previsioni nel modello di overfit siano accurate per i dati vicini ai punti dati osservati, sono molto sbagliate quando sono al di fuori di tale intervallo.

**L'overfitting è la questione più importante e impegnativa** durante la formazione di tutti i professionisti del machine learning e di tutti gli algoritmi. Come vedrai, è molto semplice creare un modello che svolga un ottimo lavoro nel fare previsioni sui dati esatti su cui è stato addestrato, ma è molto più difficile fare previsioni accurate su dati che il modello non ha mai visto prima. E, naturalmente, questi sono i dati che conteranno davvero nella pratica. Ad esempio, se crei un classificatore di cifre scritte a mano (come faremo molto presto!) e lo usi per riconoscere i numeri scritti sugli assegni, non vedrai mai nessuno dei numeri su cui è stato addestrato il modello: gli assegni avranno variazioni di scrittura leggermente diverse da gestire. In questo libro imparerai molti metodi per evitare l'adattamento eccessivo. Tuttavia, dovresti utilizzare questi metodi solo dopo aver confermato che si sta effettivamente verificando un overfitting (ovvero, hai effettivamente osservato un peggioramento dell'accuratezza della convalida durante l'addestramento). Spesso vediamo i professionisti utilizzare tecniche di evitamento dell'adattamento eccessivo anche quando dispongono di dati sufficienti di cui non avevano bisogno, finendo con un modello che potrebbe essere meno accurato di quello che avrebbero potuto ottenere.

> importante: Set di validazione: quando si addestra un modello, è necessario _sempre_ avere sia un set di addestramento che un set di validazione e misurare l'accuratezza del modello solo sul set di validazione. Se ti alleni troppo a lungo, con dati insufficienti, vedrai che la precisione del tuo modello inizierà a peggiorare; questo si chiama _overfitting_. fastai imposta `valid_pct` su `0.2` per impostazione predefinita, quindi anche se lo dimentichi, fastai creerà un set di convalida per te!

```pitone
impara = vision_learner(dls, resnet34, metrics=error_rate)
```

Perché una CNN? È l'attuale approccio all'avanguardia per la creazione di modelli di visione artificiale. Impareremo tutto su come funzionano le CNN in questo libro. La loro struttura si ispira al funzionamento del sistema visivo umano.

Esistono molte architetture diverse in fastai, che introdurremo in questo libro (oltre a discutere su come crearne una propria). Nella maggior parte dei casi, tuttavia, la scelta di un'architettura non è una parte molto importante del processo di deep learning. È qualcosa di cui gli accademici amano parlare, ma in pratica è improbabile che sia qualcosa su cui occorre dedicare molto tempo. Esistono alcune architetture standard che funzionano la maggior parte del tempo, e in questo caso ne utilizziamo una chiamata _ResNet_ di cui parleremo molto durante il libro; è veloce e preciso per molti set di dati e problemi. Il "34" in "resnet34" si riferisce al numero di livelli in questa variante dell'architettura (altre opzioni sono "18", "50", "101" e "152"). I modelli che utilizzano architetture con più livelli richiedono più tempo per l'addestramento e sono più inclini all'overfitting (ovvero non è possibile addestrarli per tante epoche prima che l'accuratezza del set di convalida inizi a peggiorare). D’altra parte, quando si utilizzano più dati, possono essere un po’ più accurati.

Cos'è una metrica? Una *metrica* è una funzione che misura la qualità delle previsioni del modello utilizzando il set di validazione e verrà stampata alla fine di ogni *epoca*. In questo caso, stiamo usando "error_rate", che è una funzione fornita da fastai che fa esattamente quello che dice: ti dice quale percentuale di immagini nel set di convalida viene classificata in modo errato. Un'altra metrica comune per la classificazione è l'"accuratezza" (che è semplicemente "1.0 - tasso_errore"). fastai ne fornisce molti altri, che verranno discussi in questo libro.

Il concetto di metrica può ricordarti la *perdita*, ma esiste un'importante distinzione. Lo scopo principale della perdita è definire una "misura della prestazione" che il sistema di allenamento può utilizzare per aggiornare automaticamente i pesi. In altre parole, una buona scelta per la perdita è una scelta facile da utilizzare per la discesa del gradiente stocastico. Ma una metrica è definita per il consumo umano, quindi una buona metrica è quella che è facile da capire e che si avvicina il più possibile a ciò che vuoi che il modello faccia. A volte potresti decidere che la funzione di perdita è una metrica adatta, ma non è necessariamente così.

"vision_learner" ha anche un parametro "pretrained", che per impostazione predefinita è "True" (quindi viene utilizzato in questo caso, anche se non lo abbiamo specificato), che imposta i pesi nel modello su valori che sono già stati addestrati da esperti per riconoscere un migliaio di categorie diverse su 1,3 milioni di foto (utilizzando il famoso [*ImageNet* dataset](http://www.image-net.org/)). Un modello che ha pesi già addestrati su qualche altro set di dati è chiamato *modello preaddestrato*. Dovresti quasi sempre utilizzare un modello preaddestrato, perché significa che il tuo modello, prima ancora di mostrargli i tuoi dati, è già molto capace. E, come vedrai, in un modello di deep learning molte di queste funzionalità sono cose di cui avrai bisogno, quasi indipendentemente dai dettagli del tuo progetto. Ad esempio, parti di modelli preaddestrati gestiranno il rilevamento di bordi, gradienti e colori, necessari per molte attività.

Quando si utilizza un modello pre-addestrato, "vision_learner" rimuoverà l'ultimo livello, poiché è sempre specificamente personalizzato per l'attività di addestramento originale (ad esempio la classificazione del set di dati ImageNet), e lo sostituirà con uno o più nuovi livelli con pesi randomizzati, di dimensioni appropriate per il set di dati con cui stai lavorando. Quest'ultima parte del modello è conosciuta come *testa*.

L'utilizzo di modelli preaddestrati è il metodo *più* importante a nostra disposizione per consentirci di addestrare modelli più accurati, più rapidamente, con meno dati e meno tempo e denaro. Potresti pensare che ciò significhi che l'uso di modelli preaddestrati sarebbe l'area più studiata nel deep learning accademico... ma ti sbaglieresti di grosso! L'importanza dei modelli preaddestrati generalmente non è riconosciuta o discussa nella maggior parte dei corsi, dei libri o delle funzionalità delle librerie software e viene raramente presa in considerazione nei documenti accademici. Mentre scriviamo questo, all’inizio del 2020, le cose stanno appena iniziando a cambiare, ma probabilmente ci vorrà del tempo. Quindi fai attenzione: la maggior parte delle persone con cui parli probabilmente sottovaluterà molto ciò che puoi fare nel deep learning con poche risorse, perché probabilmente non capiranno a fondo come utilizzare modelli preaddestrati.

L'utilizzo di un modello preaddestrato per un compito diverso da quello per cui era stato originariamente addestrato è noto come *trasferimento dell'apprendimento*. Sfortunatamente, poiché l’apprendimento basato sul trasferimento è così poco studiato, pochi ambiti dispongono di modelli preaddestrati disponibili. Ad esempio, attualmente in medicina sono disponibili pochi modelli pre-addestrati, il che rende difficile l’utilizzo del transfer learning in questo ambito. Inoltre, non è ancora ben compreso come utilizzare il transfer learning per attività quali l’analisi delle serie temporali.

> gergo: Trasferimento dell'apprendimento: utilizzo di un modello preaddestrato per un compito diverso da quello per cui era stato originariamente addestrato.

La sesta riga del nostro codice dice a fastai come *adattare* il modello:

```pitone
learn.fine_tune(1)
```

Come abbiamo discusso, l'architettura descrive solo un *modello* per una funzione matematica; in realtà non fa nulla finché non forniamo valori per i milioni di parametri che contiene.

Questa è la chiave del deep learning: determinare come adattare i parametri di un modello per far sì che risolva il problema. Per adattare un modello, dobbiamo fornire almeno un'informazione: quante volte guardare ciascuna immagine (noto come numero di *epoche*). Il numero di epoche selezionate dipenderà in gran parte da quanto tempo hai a disposizione e da quanto tempo ritieni necessario in pratica per adattare il tuo modello. Se selezioni un numero troppo piccolo, puoi sempre allenarti per più epoche in seguito.

Ma perché il metodo si chiama "fine_tune" e non "fit"? fastai in realtà *ha* un metodo chiamato `fit`, che effettivamente si adatta a un modello (cioè guarda le immagini nel set di addestramento più volte, aggiornando ogni volta i parametri per rendere le previsioni sempre più vicine alle etichette target). Ma in questo caso abbiamo iniziato con un modello pre-addestrato e non vogliamo buttare via tutte le funzionalità di cui già dispone. Come imparerai in questo libro, esistono alcuni trucchi importanti per adattare un modello preaddestrato per un nuovo set di dati, un processo chiamato *ottimizzazione*.

> gergo: Fine tuning: una tecnica di trasferimento di apprendimento in cui i parametri di un modello pre-addestrato vengono aggiornati mediante l'addestramento per epoche aggiuntive utilizzando un compito diverso da quello utilizzato per il pre-addestramento.

Quando usi il metodo "fine_tune", fastai utilizzerà questi trucchi per te. Ci sono alcuni parametri che puoi impostare (di cui parleremo più avanti), ma nel modulo predefinito mostrato qui, i passaggi sono due:

1. Utilizzare un'epoca per adattare solo le parti del modello necessarie affinché la nuova testa casuale funzioni correttamente con il set di dati.
1. Utilizzare il numero di epoche richieste quando si chiama il metodo per adattare l'intero modello, aggiornando i pesi degli strati successivi (in particolare della testa) più velocemente degli strati precedenti (che, come vedremo, generalmente non richiedono molte modifiche rispetto ai pesi preimpostati).

La *testa* di un modello è la parte che viene appena aggiunta per essere specifica del nuovo set di dati. Un'*epoca* è un passaggio completo attraverso il set di dati. Dopo aver chiamato "fit", vengono stampati i risultati dopo ogni epoca, mostrando il numero dell'epoca, le perdite del set di addestramento e convalida (la "misura delle prestazioni" utilizzata per l'addestramento del modello) e qualsiasi *metrica* richiesta (tasso di errore, in questo caso).

Quindi, con tutto questo codice il nostro modello ha imparato a riconoscere cani e gatti solo da esempi etichettati. Ma come ha fatto?

Il *machine learning* è una disciplina in cui definiamo un programma non scrivendolo interamente da soli, ma imparando dai dati. Il *deep learning* è una specialità dell'apprendimento automatico che utilizza *reti neurali* con più *livelli*. La *classificazione delle immagini* è un esempio rappresentativo (noto anche come *riconoscimento delle immagini*). Iniziamo con *dati etichettati*; cioè un insieme di immagini a cui abbiamo assegnato un'*etichetta* a ciascuna immagine indicando ciò che rappresenta. Il nostro obiettivo è produrre un programma, chiamato *modello*, che, data una nuova immagine, farà una *previsione* accurata riguardo a ciò che rappresenta quella nuova immagine.

Ogni modello inizia con la scelta dell'*architettura*, un modello generale su come funziona internamente quel tipo di modello. Il processo di *addestramento* (o *adattamento*) del modello è il processo di ricerca di un insieme di *valori di parametri* (o *pesi*) che specializzano quell'architettura generale in un modello che funzioni bene per il nostro particolare tipo di dati. Per definire il rendimento di un modello rispetto a una singola previsione, dobbiamo definire una *funzione di perdita*, che determina il modo in cui valutiamo una previsione come buona o cattiva.

Per velocizzare il processo di addestramento, potremmo iniziare con un *modello preaddestrato*, ovvero un modello che è già stato addestrato sui dati di qualcun altro. Possiamo quindi adattarlo ai nostri dati addestrandolo un po' di più sui nostri dati, un processo chiamato *fine tuning*.

Quando addestriamo un modello, una preoccupazione fondamentale è garantire che il nostro modello *generalizzi*, ovvero apprenda lezioni generali dai nostri dati che si applicano anche ai nuovi elementi che incontrerà, in modo che possa fare buone previsioni su tali elementi. Il rischio è che se addestriamo male il nostro modello, invece di imparare lezioni generali, memorizzi effettivamente ciò che ha già visto, e quindi farà previsioni inadeguate sulle nuove immagini. Un tale fallimento è chiamato *overfitting*. Per evitare ciò, dividiamo sempre i nostri dati in due parti, il *training set* e il *validation set*. Addestriamo il modello mostrandogli solo il set di training e poi valutiamo il rendimento del modello osservando le sue prestazioni sugli elementi del set di validazione. In questo modo, controlliamo se le lezioni che il modello apprende dal training set sono lezioni che si generalizzano al validation set. Affinché una persona possa valutare il rendimento complessivo del modello rispetto al set di validazione, definiamo una *metrica*. Durante il processo di addestramento, quando il modello ha visto ogni elemento del set di addestramento, lo chiamiamo *epoca*.

Tutti questi concetti si applicano all'apprendimento automatico in generale. Cioè, si applicano a tutti i tipi di schemi per definire un modello addestrandolo con i dati. Ciò che rende distintivo il deep learning è una particolare classe di architetture: le architetture basate su *reti neurali*. In particolare, compiti come la classificazione delle immagini fanno molto affidamento sulle *reti neurali convoluzionali*, di cui parleremo a breve.

## Il deep learning non serve solo per la classificazione delle immagini

L'efficacia dell'apprendimento profondo per la classificazione delle immagini è stata ampiamente discussa negli ultimi anni, mostrando anche risultati sovrumani su compiti complessi come il riconoscimento di tumori maligni nelle scansioni TC. Ma può fare molto di più, come mostreremo qui.

Parliamo, ad esempio, di qualcosa di fondamentale importanza per i veicoli autonomi: la localizzazione degli oggetti in un'immagine. Se un’auto a guida autonoma non sa dove si trova un pedone, non sa come evitarlo! La creazione di un modello in grado di riconoscere il contenuto di ogni singolo pixel in un'immagine è chiamata *segmentazione*. Ecco come possiamo addestrare un modello di segmentazione con fastai, utilizzando un sottoinsieme del [set di dati *Camvid*](http://www0.cs.ucl.ac.uk/staff/G.Brostow/papers/Brostow_2009-PRL.pdf) dal documento "Semantic Object Classes in Video: A High-Definition Ground Truth Database" di Gabruel J. Brostow, Julien Fauqueur e Roberto Cipolla:

````python
path = untar_data(URLs.CAMVID_TINY)
dls = SegmentationDataLoaders.from_label_func(
    path, bs=8, fnames = get_image_files(path/"images"),
    label_func = lambda o: path/'labels'/f'{o.stem}_P{o.suffix}',
    codes = np.loadtxt(path/'codes.txt', dtype=str)
)

learn = unet_learner(dls, resnet34)
learn.fine_tune(8)
````

> gergo: Tabulare: dati sotto forma di tabella, ad esempio da un foglio di calcolo, un database o un file CSV. Un modello tabulare è un modello che tenta di prevedere una colonna di una tabella in base alle informazioni contenute in altre colonne della tabella.

Si scopre che anche questo sembra molto simile. Ecco il codice necessario per addestrare un modello in grado di prevedere se una persona ha un reddito elevato, in base al suo background socioeconomico:

````python
from fastai.tabular.all import *
path = untar_data(URLs.ADULT_SAMPLE)

dls = TabularDataLoaders.from_csv(path/'adult.csv', path=path, y_names="salary",
    cat_names = ['workclass', 'education', 'marital-status', 'occupation',
                 'relationship', 'race'],
    cont_names = ['age', 'fnlwgt', 'education-num'],
    procs = [Categorify, FillMissing, Normalize])

learn = tabular_learner(dls, metrics=accuracy)
````

Come vedi, abbiamo dovuto dire a fastai quali colonne sono *categoriche* (ovvero contengono valori che fanno parte di un insieme discreto di scelte, come "occupazione") e quali sono *continue* (ovvero contengono un numero che rappresenta una quantità, come "età").

Non è disponibile un modello preaddestrato per questa attività (in generale, i modelli preaddestrati non sono ampiamente disponibili per le attività di modellazione tabulare, sebbene alcune organizzazioni li abbiano creati per uso interno), quindi in questo caso non utilizziamo "fine_tune". Utilizziamo invece `fit_one_cycle`, il metodo più comunemente utilizzato per addestrare i modelli fastai *da zero* (ovvero senza trasferimento di apprendimento):

````python
learn.fit_one_cycle(3)
````

Questo modello utilizza il [set di dati*Adult*](http://robotics.stanford.edu/~ronnyk/nbtree.pdf), dal documento "Scaling Up the Accuracy of Naive-Bayes Classifiers: a Decision-Tree Hybrid" di Rob Kohavi, che contiene alcuni dati demografici sugli individui (come la loro istruzione, stato civile, razza, sesso e se hanno o meno un reddito annuo superiore a $ 50.000). Il modello ha una precisione superiore all'80\% e ha impiegato circa 30 secondi per l'addestramento.

## Set di validazione e set di test

Come abbiamo discusso, l'obiettivo di un modello è fare previsioni sui dati. Ma il processo di formazione del modello è fondamentalmente stupido. Se addestrassimo un modello con tutti i nostri dati e poi valutassimo il modello utilizzando gli stessi dati, non saremmo in grado di dire quanto bene il nostro modello può funzionare sui dati che non ha visto. Senza questa preziosa informazione che ci guida nell'addestramento del nostro modello, ci sono ottime possibilità che diventi bravo a fare previsioni su tali dati ma otterrebbe prestazioni scarse con i nuovi dati.

Per evitare ciò, il nostro primo passo è stato quello di dividere il nostro set di dati in due set: il *set di addestramento* (che il nostro modello vede nell'addestramento) e il *set di validazione*, noto anche come *set di sviluppo* (che viene utilizzato solo per la valutazione). Ciò ci consente di verificare che il modello apprende lezioni dai dati di training che si generalizzano a nuovi dati, i dati di convalida.

Un modo per comprendere questa situazione è che, in un certo senso, non vogliamo che il nostro modello ottenga buoni risultati “imbrogliando”. Se fa una previsione accurata per un elemento di dati, dovrebbe essere perché ha appreso le caratteristiche di quel tipo di elemento e non perché il modello è stato modellato *avendo effettivamente visto quel particolare elemento*.

Separare i nostri dati di convalida significa che il nostro modello non li vede mai durante l'addestramento e quindi ne è completamente incontaminato e non imbroglia in alcun modo. Giusto?

In effetti, non necessariamente. La situazione è più sottile. Questo perché in scenari realistici raramente costruiamo un modello semplicemente addestrandone i parametri di peso una volta. Invece, è probabile che esploreremo molte versioni di un modello attraverso varie scelte di modellazione riguardanti l'architettura di rete, i tassi di apprendimento, le strategie di aumento dei dati e altri fattori di cui parleremo nei prossimi capitoli. Molte di queste scelte possono essere descritte come scelte di *iperparametri*. La parola riflette che si tratta di parametri su parametri, poiché sono le scelte di livello superiore che governano il significato dei parametri di peso.

Il problema è che anche se il normale processo di allenamento esamina solo le previsioni sui dati di allenamento quando apprende i valori per i parametri di peso, lo stesso non vale per noi. Noi, come modellatori, stiamo valutando il modello esaminando le previsioni sui dati di validazione quando decidiamo di esplorare nuovi valori di iperparametri! Quindi le versioni successive del modello vengono, indirettamente, modellate da noi dopo aver visto i dati di validazione. Proprio come il processo di addestramento automatico rischia di sovraccaricare i dati di addestramento, noi corriamo il rischio di sovraccaricare i dati di validazione attraverso tentativi ed errori umani ed esplorazioni.

La soluzione a questo enigma è introdurre un altro livello di dati ancora più riservati, il *test set*. Proprio come tratteniamo i dati di validazione dal processo di formazione, dobbiamo trattenere i dati del set di test anche da noi stessi. Non può essere utilizzato per migliorare il modello; può essere utilizzato solo per valutare il modello alla fine dei nostri sforzi. In effetti, definiamo una gerarchia di tagli dei nostri dati, in base a quanto completamente vogliamo nasconderli dai processi di training e modellazione: i dati di training sono completamente esposti, i dati di validazione sono meno esposti e i dati di test sono totalmente nascosti. Questa gerarchia è parallela ai diversi tipi di processi di modellazione e valutazione stessi: il processo di addestramento automatico con propagazione all'indietro, il processo più manuale di prova di diversi iperparametri tra le sessioni di addestramento e la valutazione del nostro risultato finale.

I set di test e convalida dovrebbero contenere dati sufficienti per garantire una buona stima della precisione. Se stai creando un rilevatore di gatti, ad esempio, generalmente desideri almeno 30 gatti nel tuo set di convalida. Ciò significa che se disponi di un set di dati con migliaia di elementi, l'utilizzo della dimensione del set di convalida predefinita del 20% potrebbe essere superiore a quella necessaria. D'altra parte, se si dispone di molti dati, utilizzarne una parte per la convalida probabilmente non presenta svantaggi.

Avere due livelli di "dati riservati" (un set di convalida e un set di test, con un livello che rappresenta i dati che virtualmente nascondi a te stesso) può sembrare un po' estremo. Ma il motivo per cui è spesso necessario è perché i modelli tendono a gravitare verso il modo più semplice per fare buone previsioni (memorizzazione), e noi, come esseri umani fallibili, tendiamo a gravitare verso illuderci sulle prestazioni dei nostri modelli. La disciplina del test ci aiuta a mantenerci intellettualmente onesti. Ciò non significa che abbiamo *sempre* bisogno di un set di test separato: se hai pochissimi dati, potrebbe essere necessario avere solo un set di convalida, ma generalmente è meglio usarne uno se possibile.

Questa stessa disciplina può essere fondamentale se intendi assumere una terza parte per eseguire lavori di modellazione per tuo conto. Una terza parte potrebbe non comprendere accuratamente le tue esigenze o i suoi incentivi potrebbero addirittura incoraggiarli a fraintenderli. Un buon set di test può mitigare notevolmente questi rischi e consentirti di valutare se il loro lavoro risolve il tuo problema reale.

Per dirla senza mezzi termini, se sei un decisore senior nella tua organizzazione (o fornisci consulenza a decisori senior), il punto più importante è questo: se ti assicuri di comprendere veramente cosa sono i set di test e validazione e perché sono importanti, allora eviterai la più grande fonte di fallimenti che abbiamo visto quando le organizzazioni decidono di utilizzare l'intelligenza artificiale. Ad esempio, se stai pensando di coinvolgere un fornitore o un servizio esterno, assicurati di fornire alcuni dati di test che il fornitore *non vedrà mai*. Quindi *tu* controlli il loro modello sui dati del tuo test, utilizzando una metrica che *tu* scegli in base a ciò che conta realmente per te nella pratica, e *tu* decidi quale livello di prestazione è adeguato. (È anche una buona idea provare tu stesso alcune semplici linee di base, in modo da sapere cosa può ottenere un modello veramente semplice. Spesso si scopre che il tuo modello semplice funziona altrettanto bene di quello prodotto da un "esperto" esterno!)

### Utilizzare il giudizio nella definizione dei set di test

Per fare un buon lavoro nel definire un set di validazione (e possibilmente un set di test), a volte vorrai fare di più che semplicemente prendere casualmente una frazione del tuo set di dati originale. Ricorda: una proprietà chiave dei set di validazione e test è che devono essere rappresentativi dei nuovi dati che vedrai in futuro. Potrebbe sembrare un ordine impossibile! Per definizione, non hai ancora visto questi dati. Ma di solito sai ancora alcune cose.

È istruttivo osservare alcuni casi di esempio. Molti di questi esempi provengono da concorsi di modellazione predittiva sulla piattaforma [Kaggle](https://www.kaggle.com/), che è una buona rappresentazione dei problemi e dei metodi che potresti riscontrare nella pratica.

Un caso potrebbe essere se stai guardando i dati delle serie temporali. Per una serie temporale, scegliere un sottoinsieme casuale di dati sarà troppo semplice (puoi guardare i dati sia prima che dopo le date che stai cercando di prevedere) e non rappresentativo della maggior parte dei casi d'uso aziendali (dove stai utilizzando dati storici per costruire un modello da utilizzare in futuro). Se i tuoi dati includono la data e stai creando un modello da utilizzare in futuro, ti consigliamo di scegliere una sezione continua con le date più recenti come set di convalida (ad esempio, le ultime due settimane o l'ultimo mese di dati disponibili).
